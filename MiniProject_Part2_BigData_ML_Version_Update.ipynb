## 1. ปัญหาและคำถามที่สนใจ
**ปัญหา/คำถามที่สนใจคืออะไร** <br>
ปัจจัยด้าน`สภาพอากาศ` `เวลา` `สถานที่` และ`ลักษณะถนน` สามารถใช้ทำนายความรุนแรงของอุบัติเหตุทางถนนในสหรัฐอเมริกาได้หรือไม่ และปัจจัยใดมีผลต่อความรุนแรงมากที่สุด<br><br>

**ปัญหา/คำถามนี้มีความสำคัญอย่างไร และใครจะนำผลไปใช้** <br>
ความสำคัญ:

- อุบัติเหตุทางถนนก่อให้เกิดการบาดเจ็บ เสียชีวิต และความเสียหายทางเศรษฐกิจจำนวนมากทุกปี
- หากทำนายได้ว่าอุบัติเหตุแบบใดมีแนวโน้มรุนแรง จะช่วยจัดสรรทรัพยากรและวางมาตรการป้องกันได้ตรงจุดมากขึ้น
- ข้อมูลมีขนาดประมาณ 7.7 ล้านแถว จึงเหมาะกับการใช้เทคโนโลยี Big Data (PySpark) ในการประมวลผล <br><br>
ผู้ใช้ประโยชน์:

- หน่วยงานดูแลความปลอดภัยทางถนนและผู้วางแผนการจราจร ใช้กำหนดจุดเสี่ยงและช่วงเวลาที่ต้องเฝ้าระวัง
- หน่วยแพทย์ฉุกเฉินและตำรวจทางหลวง ใช้ประเมินความต้องการทรัพยากรล่วงหน้า
- บริษัทประกันภัย ใช้ประกอบการประเมินความเสี่ยง
- แอปนำทางและผู้ขับขี่ ใช้เตือนเส้นทางหรือสภาพอากาศที่เสี่ยงสูง<br><br>

**เป้าหมาย (target) และประเภทงาน (classification / regression / unsupervised)**<br>
- **Target:** `Severity` ในชุดข้อมูลเดิมมี 4 ระดับ (1–4) แนะนำให้แปลงเป็น binary เพื่อให้โจทย์ชัดและลดปัญหาข้อมูลไม่สมดุล เช่น ระดับ 1–2 = ไม่รุนแรง (0) และระดับ 3–4 = รุนแรง (1) ควรตรวจสัดส่วนของแต่ละระดับในข้อมูลจริงก่อนตัดสินใจเกณฑ์แบ่ง
- **ประเภทงาน:** Supervised Learning แบบ **Classification** (Binary Classification)
- **โมเดลที่จะเปรียบเทียบ:** Logistic Regression, Random Forest, Gradient-Boosted Trees (Spark MLlib)
- **ตัวชี้วัด:** F1-score, AUC-ROC, Precision/Recall (ไม่ใช้ Accuracy อย่างเดียว เพราะข้อมูลอาจไม่สมดุล)

### 2.4 Data Dictionary
**ตัวแปรตาม และตัวแปรสำหรับตรวจข้อมูล**
| คอลัมน์ | ความหมาย | บทบาท |
|---|---|---|
| `Severity` | ระดับผลกระทบต่อการจราจร (1 = น้อยที่สุด, 4 = มากที่สุด) | **Target** แปลงเป็น `y_severe` = 1 เมื่อ Severity ≥ 3 (ระดับ 3–4) |
| `Source` | แหล่งที่รายงานอุบัติเหตุ (Source1–3) | ใช้ตรวจคุณภาพข้อมูลและกรองช่วงข้อมูล [ถ้าตัดสินใจ] ไม่ใช้เป็น feature (เป็นร่องรอยวิธีเก็บข้อมูล ไม่ใช่ลักษณะอุบัติเหตุ) |
| `ID` | รหัสเฉพาะของแต่ละเหตุการณ์ | ไม่ใช้ |

**กลุ่มสถานที่**
| คอลัมน์ | ความหมาย | บทบาท |
|---|---|---|

| `Description` | คำอธิบายเหตุการณ์ | ไม่ใช้ (อาจมีข้อมูลหลังเกิดเหตุ) |
| `Street`, `City`, `County`, `Zipcode`, `Airport_Code` | ที่อยู่/รหัส | ไม่ใช้ (ค่าไม่ซ้ำสูงมาก) |

| คอลัมน์ | ความหมาย | บทบาท |

### 2.5 ข้อสังเกตสำคัญของข้อมูล (รายละเอียดในข้อ 4 และ 6)

```python
os.makedirs(output_dir, exist_ok=True)
chunksize = 100_000
for i, chunk in enumerate(
    pd.read_csv(input_file, chunksize=chunksize)
):
    output_file = os.path.join(
        output_dir, f"part_{i:04d}.parquet"
    )
    chunk.to_parquet(
        output_file,
        engine="pyarrow",
        compression="snappy",
        index=False
    )
    print(f"Chunk {i}: {len(chunk):,} rows saved")
    del chunk
    gc.collect()
print("Conversion completed!")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu

# 1. โหลดข้อมูล Severity และตัวแปรด้านสภาพอากาศที่สำคัญ
analysis_cols = ['Severity', 'Temperature(F)', 'Humidity(%)', 'Visibility(mi)', 'Wind_Speed(mph)', 'Precipitation(in)']
df_stat = dataset.to_table(columns=analysis_cols).to_pandas()

# 2. แบ่งกลุ่มความรุนแรง: 1-2 คือ 'น้อย (Low)' และ 3-4 คือ 'รุนแรง (Severe)'
df_stat['Severity_Group'] = np.where(df_stat['Severity'].isin([1, 2]), 'Low', 'Severe')

group_low = df_stat[df_stat['Severity_Group'] == 'Low']
group_severe = df_stat[df_stat['Severity_Group'] == 'Severe']

print(f"จำนวนอุบัติเหตุกลุ่มรุนแรงน้อย (Severity 1-2): {len(group_low):,} เคส")
print(f"จำนวนอุบัติเหตุกลุ่มรุนแรงมาก (Severity 3-4): {len(group_severe):,} เคส\n")

# 3. รัน Mann-Whitney U Test สำหรับตัวแปรสภาพอากาศหลักๆ
results = []
variables = ['Temperature(F)', 'Humidity(%)', 'Visibility(mi)', 'Wind_Speed(mph)']

print("=== ผลการทดสอบ Mann-Whitney U Test ===")
for var in variables:
    # ตัดค่า Missing (NaN) ออกสำหรับการทดสอบ
    data_low = group_low[var].dropna()
    data_severe = group_severe[var].dropna()

    # คำนวณ Mann-Whitney U Test
    stat, p_val = mannwhitneyu(data_low, data_severe, alternative='two-sided')

    # สรุปค่ามัธยฐานของแต่ละกลุ่ม
    median_low = data_low.median()
    median_severe = data_severe.median()

    # การสรุปผล
    sig = "มีความแตกต่างกันอย่างมีนัยสำคัญทางสถิติ (p < 0.05)" if p_val < 0.05 else "ไม่มีความแตกต่างกันอย่างมีนัยสำคัญทางสถิติ"

    print(f"\nตัวแปร: {var}")
    print(f"  - ค่ามัธยฐานกลุ่มรุนแรงน้อย (Low): {median_low}")
    print(f"  - ค่ามัธยฐานกลุ่มรุนแรงมาก (Severe): {median_severe}")
    print(f"  - U-statistic: {stat:,.1f} | p-value: {p_val:.3e}")
    print(f"  - สรุปผล: {sig}")
```

```python
######################################################################
# CELL 4: ค่าที่เป็นไปไม่ได้
######################################################################
# 5.3 ค่าที่เป็นไปไม่ได้ -> เปลี่ยนเป็นค่าว่าง (NaN) ไม่ลบทั้งแถว
CLEAN_RULES = {
    "Temperature(F)": (-80, 135),
    "Humidity(%)": (0, 100),
    "Pressure(in)": (20, 33),      # ผ่อนจาก 25 เพราะพื้นที่สูงมีความกดต่ำกว่า (พบ 96,846 ค่าในช่วง 20-25)
    "Visibility(mi)": (0, 100),
    "Wind_Speed(mph)": (0, 250),
    "Precipitation(in)": (0, 25),
}
range_before = raw_df[list(CLEAN_RULES)].agg(["min", "max"]).T

detail, total_changed = [], 0
for col, (lo, hi) in CLEAN_RULES.items():
    bad = (raw_df[col] < lo) | (raw_df[col] > hi)
    n_bad = int(bad.sum())
    raw_df.loc[bad, col] = np.nan
    total_changed += n_bad
    detail.append({"column": col, "ช่วงที่ยอมรับ": f"{lo} ถึง {hi}", "เปลี่ยนเป็นค่าว่าง": n_bad})
display(pd.DataFrame(detail))
log_step("3. ค่าเป็นไปไม่ได้", "เปลี่ยนค่านอกช่วงที่เป็นไปได้เป็น NaN",
         "เป็นข้อผิดพลาดของการวัด/บันทึก; ไม่ลบแถวเพราะคอลัมน์อื่นยังใช้ได้", f"{total_changed:,} ช่อง", len(raw_df))
```

```python
######################################################################
# CELL 5: หมวดหมู่ที่ไม่สอดคล้อง
######################################################################
# 5.4 รวมค่าหมวดหมู่ที่หมายถึงอย่างเดียวกัน (Wind_Direction)
def remap_categories(s, func):
    """รวม/เปลี่ยนชื่อหมวดหมู่ โดยทำงานบนรายการ categories (เร็ว ไม่วนทีละแถว)"""
    cats = list(s.cat.categories)
    new_labels = [func(c) for c in cats]
    new_cats = list(dict.fromkeys(new_labels))
    pos = {c: i for i, c in enumerate(new_cats)}
    code_map = np.array([pos[x] for x in new_labels], dtype=np.int32)
    codes = s.cat.codes.to_numpy()
    new_codes = np.where(codes >= 0, code_map[np.clip(codes, 0, None)], -1)
    return pd.Series(pd.Categorical.from_codes(new_codes, categories=new_cats), index=s.index)

def wind_dir_std(x):
    m = {"north": "N", "south": "S", "east": "E", "west": "W",
         "calm": "CALM", "variable": "VAR", "var": "VAR"}
    t = str(x).strip().lower()
    return m.get(t, t.upper())

n_before = raw_df["Wind_Direction"].nunique()
raw_df["Wind_Direction"] = remap_categories(raw_df["Wind_Direction"], wind_dir_std)
n_after = raw_df["Wind_Direction"].nunique()
print(f"Wind_Direction: {n_before} ค่า -> {n_after} ค่า")
print(sorted(raw_df["Wind_Direction"].cat.categories))
log_step("4. หมวดหมู่ไม่สอดคล้อง", "รวมค่า Wind_Direction ที่ความหมายเดียวกัน (North→N, Calm→CALM, Variable→VAR ฯลฯ)",
         "ป้องกันโมเดลมองเป็นคนละกลุ่ม", f"{n_before} -> {n_after} ค่า", len(raw_df))
```

```python
######################################################################
# CELL 6: Outliers
######################################################################
# 5.5 Outliers: เทียบช่วงค่าก่อน/หลังตัดค่าเป็นไปไม่ได้ และนับ outlier ที่เหลือ (IQR)
range_after = raw_df[list(CLEAN_RULES)].agg(["min", "max"]).T
display(range_before.join(range_after, lsuffix="_ก่อน", rsuffix="_หลัง").round(2))

num_cols = ["Temperature(F)", "Humidity(%)", "Pressure(in)", "Visibility(mi)", "Wind_Speed(mph)", "Precipitation(in)"]
q = raw_df[num_cols].quantile([0.25, 0.75])
iqr = q.loc[0.75] - q.loc[0.25]
lower, upper = q.loc[0.25] - 1.5 * iqr, q.loc[0.75] + 1.5 * iqr
out = [{"column": c, "IQR": round(iqr[c], 3),
        "outliers (IQR)": int(((raw_df[c] < lower[c]) | (raw_df[c] > upper[c])).sum())} for c in num_cols]
display(pd.DataFrame(out))
log_step("5. Outliers ทางสถิติ", "เก็บไว้ (ไม่ลบ ไม่ปรับค่า)",
         "หลังตัดค่าเป็นไปไม่ได้แล้ว ที่เหลือคือสภาพอากาศสุดโต่งที่เกิดขึ้นจริง และอาจสัมพันธ์กับความรุนแรง",
         "0", len(raw_df))
```

   - **ผลต่อการตัดสินใจ:** (ก) ไม่ใช้ปี (`year`) เป็น feature (ข) เทียบความสัมพันธ์ของปัจจัยภายในปีเดียวกัน (ค) พิจารณาไม่ใช้ตัวบอก "เคยหาย" ของ `Precipitation`/`Wind_Speed` เพราะอาจกลายเป็นตัวแทนของปี (ง) การแบ่ง train/validation/test ต้องคำนึงถึงแนวโน้มนี้ เพราะสัดส่วนเหตุรุนแรงของช่วงต้นกับท้ายต่างกันมาก และรายงานผลเทียบกับค่าเดาของแต่ละชุดหรือแยกรายปี (จ) ตัดสินใจเลือกช่วงข้อมูลที่ป้ายกำกับสม่ำเสมอ [ใส่ผลที่กลุ่มตกลง]

3. **ความสัมพันธ์ของปัจจัยกับความรุนแรง (เทียบภายในปีเดียวกัน ใช้ Cramér's V)** สถานที่ (State, 0.25) สัมพันธ์สูงสุด รองลงมาคือลักษณะถนน (Traffic_Signal 0.13, Crossing 0.12) และเวลา (เดือน 0.09, ชั่วโมง 0.08, วันในสัปดาห์ 0.07) ส่วนสภาพอากาศอ่อนที่สุด (Weather condition 0.05) โดยตัวแปรอากาศที่ดูสัมพันธ์สูงเมื่อรวมทุกปี (0.20) อ่อนลงมากเมื่อตัดแนวโน้มรายปี อย่างไรก็ตามปัจจัยเดี่ยวสัมพันธ์ระดับไม่สูง (V สูงสุดเพียง 0.25)
   - **ผลต่อการตัดสินใจ:** สร้าง feature จากทั้งสี่กลุ่ม (ชั่วโมง วันในสัปดาห์ ช่วงเร่งด่วน State พิกัด ลักษณะถนน จัดกลุ่ม `Weather_Condition`) ใช้โมเดลที่รวมหลายปัจจัยพร้อมกัน (Logistic Regression, Random Forest, HistGradientBoosting) และให้ผลโมเดลตัดสินว่าทำนายได้ดีกว่าการเดาเพียงใด ระวังตัวแปรเดือน เพราะปี 2016 เริ่มกลางเดือนมกราคมและปี 2023 มีข้อมูลแค่ 3 เดือน

**ข้อจำกัด:** ทั้งหมดเป็นความสัมพันธ์ ไม่ใช่เหตุและผล, State อาจสะท้อนวิธีบันทึกที่ต่างกันตามภูมิภาค, และข้อมูลรวมจากหลายแหล่งที่ให้ระดับ Severity ไม่เหมือนกัน

| ลักษณะถนน 12 ตัว | คอลัมน์เดิม | Traffic_Signal 0.13, Crossing 0.12 สูงและไม่ผันตามปี | ใช่ |
| `Temperature`, `Humidity`, `Pressure`, `Visibility`, `Wind_Direction` | คอลัมน์เดิม | องค์ประกอบของกลุ่มสภาพอากาศ | ใช่ |
| `Wind_Speed`, `Precipitation` | คอลัมน์เดิม | **ยังไม่ใช้เป็นค่าเริ่มต้น**: ว่างเกือบหมดในปีแรก ๆ (Precipitation 87–91%, Wind_Speed ~19% ในปี 2016–2018) อาจกลายเป็นตัวแทนของปี ตัดสินใจหลังเลือกช่วงข้อมูล | ใช่ |

**ไม่ใช้:** `End_Time`, `Distance(mi)`, `End_Lat/Lng`, `Description` (รู้หลังเกิดเหตุ), `ID`, `year` (ปีในอนาคตที่โมเดลไม่เคยเห็น และเป็นตัวแทนของวิธีเก็บข้อมูล), `Source` (ไม่ใช่ลักษณะของอุบัติเหตุ)

**ข้อสังเกต:** ตัวแปรเดือนอาจปนกับแนวโน้มปี เพราะปี 2016 เริ่มกลางเดือนมกราคม และปี 2023 มีข้อมูลแค่ 3 เดือน

```python
# ข้อ 7: Feature Engineering
import numpy as np
import pandas as pd

df = clean_df
SEVERE_MIN = 3
df["y_severe"] = (df["Severity"] >= SEVERE_MIN).astype("int8")      # target: 1 = รุนแรง (ระดับ 3-4)

# ---- ตัวเลือกที่ต้องตัดสินใจร่วมกับกลุ่ม ----
# Precipitation/Wind_Speed ว่างเกือบหมดในปีแรก ๆ (87-91% / 19%) ถ้าใส่เป็น feature อาจกลายเป็น "ตัวแทนของปี" แทนที่จะเป็นอากาศจริง
# ตั้ง False ไว้ก่อน เปลี่ยนเป็น True ได้ถ้าเลือกช่วงข้อมูลที่ค่าเหล่านี้ครบ (เช่น ปี 2020 เป็นต้นไป)
INCLUDE_RISKY_WEATHER = False

# ---- 1) ฟีเจอร์เวลา (จาก Start_Time) ----
t = df["Start_Time"]
df["hour"] = t.dt.hour.astype("int8")
df["weekday"] = t.dt.weekday.astype("int8")            # 0 = จันทร์
df["month"] = t.dt.month.astype("int8")
df["is_weekend"] = df["weekday"] >= 5
df["is_rush_hour"] = (~df["is_weekend"]) & df["hour"].isin([7, 8, 9, 16, 17, 18])

# ---- 2) กลางวัน/กลางคืน (ค่าว่างเป็น "Unknown") ----
df["day_night"] = df["Sunrise_Sunset"].astype("category")
if "Unknown" not in df["day_night"].cat.categories:
    df["day_night"] = df["day_night"].cat.add_categories("Unknown")
df["day_night"] = df["day_night"].fillna("Unknown")

# ---- 3) จัดกลุ่มสภาพอากาศ (Weather_Condition หลายสิบ-ร้อยค่า -> 8 กลุ่ม) ----
def remap_categories(s, func):
    """รวม/เปลี่ยนชื่อหมวดหมู่ด้วยฟังก์ชัน โดยทำงานบนรายการ categories (เร็ว ไม่วนทีละแถว)"""
    cats = list(s.cat.categories)
    new_labels = [func(c) for c in cats]
    new_cats = list(dict.fromkeys(new_labels))
    pos = {c: i for i, c in enumerate(new_cats)}
    code_map = np.array([pos[x] for x in new_labels], dtype=np.int32)
    codes = s.cat.codes.to_numpy()
    new_codes = np.where(codes >= 0, code_map[np.clip(codes, 0, None)], -1)
    return pd.Series(pd.Categorical.from_codes(new_codes, categories=new_cats), index=s.index)

def weather_group(x):
    s = str(x).lower()
    if "thunder" in s or "t-storm" in s:
        return "Thunderstorm"
    if any(k in s for k in ["snow", "sleet", "wintry", "ice", "freezing", "hail", "blizzard"]):
        return "Snow/Ice"
    if any(k in s for k in ["rain", "drizzle", "shower", "precipitation"]):
        return "Rain"
    if any(k in s for k in ["fog", "mist", "haze", "smoke", "dust", "sand", "ash"]):
        return "Fog/Haze/Smoke"
    if any(k in s for k in ["cloud", "overcast"]):
        return "Cloudy"
    if any(k in s for k in ["clear", "fair"]):
        return "Clear"
    if any(k in s for k in ["wind", "squall", "tornado", "funnel"]):
        return "Windy/Other"
    return "Other"

df["Weather_Group"] = remap_categories(df["Weather_Condition"], weather_group)
for c in ["Weather_Group", "Wind_Direction"]:
    if "Unknown" not in df[c].cat.categories:
        df[c] = df[c].cat.add_categories("Unknown")
    df[c] = df[c].fillna("Unknown")      # ค่าว่าง = กลุ่ม "Unknown"

# ---- 4) จำนวนลักษณะถนนที่พบรวมกัน (aggregation) ----
ROAD_COLS = ["Amenity", "Bump", "Crossing", "Give_Way", "Junction", "No_Exit", "Railway",
             "Roundabout", "Station", "Stop", "Traffic_Calming", "Traffic_Signal"]
df["n_road_features"] = df[ROAD_COLS].sum(axis=1).astype("int8")

# ---- รายการ feature สุดท้าย ----
CAT_COLS = ["State", "Wind_Direction", "Weather_Group", "day_night", "hour", "weekday", "month"]
NUM_COLS = ["Temperature(F)", "Humidity(%)", "Pressure(in)", "Visibility(mi)", "Start_Lat", "Start_Lng", "n_road_features"]
if INCLUDE_RISKY_WEATHER:
    NUM_COLS += ["Wind_Speed(mph)", "Precipitation(in)"]
BIN_COLS = ROAD_COLS + ["is_weekend", "is_rush_hour"]
FEATURES = CAT_COLS + NUM_COLS + BIN_COLS

# ---- ตรวจ data leakage: ห้ามมีคอลัมน์ที่รู้หลังเกิดเหตุ หรือที่ไม่ควรใช้ ----
FORBIDDEN = {"End_Time", "Distance(mi)", "End_Lat", "End_Lng", "Description", "ID", "Severity", "y_severe",
             "Source", "Start_Time", "Weather_Timestamp", "year"}
leak = FORBIDDEN.intersection(FEATURES)
assert not leak, f"พบคอลัมน์ต้องห้ามใน feature: {leak}"

# ---- ตารางสำหรับขั้นต่อไป (เก็บ Start_Time ไว้แบ่งข้อมูล และ Severity/y_severe เป็น target ไม่ใช่ feature) ----
model_df = df[FEATURES + ["y_severe", "Severity", "Start_Time"]].copy()
# ---- แสดงผลลัพธ์ ----
print(f"model_df: {model_df.shape[0]:,} แถว | features = {len(FEATURES)} "
      f"(หมวดหมู่ {len(CAT_COLS)}, ตัวเลข {len(NUM_COLS)}, 0/1 {len(BIN_COLS)}) | {model_df.memory_usage(deep=True).sum() / 1e9:.2f} GB")
print("หมวดหมู่:", CAT_COLS)
print("ตัวเลข  :", NUM_COLS)
print("0/1     :", BIN_COLS)
print("ตรวจ leakage: ผ่าน (ไม่มีคอลัมน์ต้องห้ามใน feature)")

print("\nสัดส่วนกลุ่มสภาพอากาศ (Weather_Group):")
display((df["Weather_Group"].value_counts(normalize=True) * 100).round(2).to_frame("% ของแถว"))

print("ตัวอย่าง feature ใหม่ 5 แถวแรก:")
display(model_df[["Start_Time", "hour", "weekday", "month", "is_weekend", "is_rush_hour", "day_night", "Weather_Group", "n_road_features", "y_severe"]].head())

# ตรวจความซ้ำซ้อนของตัวแปรตัวเลข (Spearman บนตัวอย่าง 300,000 แถว)
cs = df[NUM_COLS].sample(n=min(300_000, len(df)), random_state=42).corr(method="spearman")
pairs = cs.where(np.triu(np.ones(cs.shape, dtype=bool), 1)).stack()
print("คู่ตัวเลขที่สัมพันธ์กันเองสูงสุด:", f"{pairs.abs().idxmax()} |rho| = {pairs.abs().max():.2f}",
      "-> ต่ำกว่า 0.95 ไม่ต้องตัดตัวใด" if pairs.abs().max() < 0.95 else "-> ควรตัดตัวหนึ่ง")
```

## 8. Data Partitioning
อธิบายวิธีแบ่ง train / validation / test และเหตุผลที่วิธีนี้เหมาะกับข้อมูลและคำถามของกลุ่ม

กลยุทธ์การแบ่งข้อมูลที่เหมาะสมที่สุดคือ Time-Based Split (การแบ่งข้อมูลตามลำดับเวลา) หรือ Out-of-Time (OOT) Validation แทนการสุ่มแบ่งแบบปกติ (Random Split)

**การแบ่งข้อมูล (Time-Based Split)** เรียงลำดับข้อมูลตามคอลัมน์ Start_Time จากอดีตไปหาปัจจุบัน จากนั้นตัดแบ่งข้อมูลตามช่วงเวลา โดยอิงจากสัดส่วนของข้อมูลหรือจำนวนปี/เดือนที่มี

**โดยมีสัดส่วนดังนี้:**   
- Training Set (70%): ใช้ข้อมูลช่วงแรกหรือข้อมูลช่วงที่เก่าที่สุด เพื่อให้โมเดลเรียนรู้รูปแบบความสัมพันธ์ของสภาพอากาศ และปัจจัยแวดล้อมที่ส่งผลต่อความรุนแรงของอุบัติเหตุ
- Validation Set (15%): ใช้ข้อมูลช่วงถัดมา สำหรับปรับจูน Hyperparameter และเลือกโมเดลที่ดีที่สุด
- Test Set (15%): ใช้ข้อมูลช่วงล่าสุด (ใหม่ที่สุด) เพื่อประเมินประสิทธิภาพขั้นสุดท้ายของโมเดลก่อนนำไปใช้งานจริง

```python
# Partition data
import pandas as pd
import numpy as np
# 1. เรียงลำดับข้อมูลตาม 'Start_Time' จากอดีตไปปัจจุบันเพื่อป้องกัน Data Leakage
model_df = model_df.sort_values('Start_Time').reset_index(drop=True)

# 2. กำหนดสัดส่วน Train 70%, Validation 15%, Test 15%
total_rows = len(model_df)
train_end = int(total_rows * 0.70)
val_end = int(total_rows * 0.85)

# 3. แบ่งข้อมูลตามลำดับเวลา
train_df = model_df.iloc[:train_end]
val_df = model_df.iloc[train_end:val_end]
test_df = model_df.iloc[val_end:]

# 4. แยก Features (X) และ Target (y)
cols_to_drop = ['y_severe', 'Severity', 'Start_Time']

X_train = train_df.drop(columns=cols_to_drop)
y_train = train_df['y_severe']

X_val = val_df.drop(columns=cols_to_drop)
y_val = val_df['y_severe']

X_test = test_df.drop(columns=cols_to_drop)
y_test = test_df['y_severe']

# 5. ตรวจสอบผลลัพธ์
print("สรุปขนาดของชุดข้อมูลหลัง Partition:")
print(f"Train Shape: {X_train.shape} | สัดส่วน: {len(X_train)/total_rows:.1%}")
print(f"Val Shape:   {X_val.shape} | สัดส่วน: {len(X_val)/total_rows:.1%}")
print(f"Test Shape:  {X_test.shape} | สัดส่วน: {len(X_test)/total_rows:.1%}")
```

## 9. Preprocessing Pipeline
การจัดการ missing values, encoding, scaling ที่ต้อง "เรียนรู้จากข้อมูล" ให้ fit จาก **train set เท่านั้น**
แล้วนำไป transform ชุดอื่น (แนะนำใช้ `sklearn.pipeline.Pipeline` / `ColumnTransformer`)

```python
# Build preprocessing pipeline
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import pandas as pd

num_cols = NUM_COLS
cat_cols = CAT_COLS
bool_cols = BIN_COLS

# 2. แก้ปัญหาชนิดข้อมูล: แปลงคอลัมน์กลุ่ม bool_cols ให้เป็นตัวเลข int (0/1)
for col in bool_cols:
    X_train[col] = X_train[col].astype(int)
    X_val[col] = X_val[col].astype(int)
    X_test[col] = X_test[col].astype(int)

# 3. สร้าง Pipeline ย่อยสำหรับแต่ละประเภทข้อมูล
# 3.1 ข้อมูลตัวเลข (Numerical)
num_pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())])

# 3.2 ข้อมูลหมวดหมู่ (Categorical)
cat_pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='most_frequent')),
    # ใช้ sparse_output=True เพื่อประหยัด RAM บนข้อมูลขนาดใหญ่
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=True)) ])

# 3.3 ข้อมูล 0/1 (Binary)
bool_pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='most_frequent'))])

# 4. รวม Pipeline ย่อยทั้งหมดเข้าด้วยกัน
preprocessor = ColumnTransformer(transformers=[('num', num_pipeline, num_cols),
        ('cat', cat_pipeline, cat_cols),('bool', bool_pipeline, bool_cols)],remainder='drop' )

# 5. Fit บน Train Set และ Transform (เรียนรู้จาก Train และนำไปประยุกต์ใช้กับทุกชุด)
X_train_processed = preprocessor.fit_transform(X_train)
X_val_processed = preprocessor.transform(X_val)
X_test_processed = preprocessor.transform(X_test)

print("สรุปขนาดของข้อมูลหลังผ่าน Preprocessing Pipeline:")
print(f"X_train_processed shape: {X_train_processed.shape}")
print(f"X_val_processed shape:   {X_val_processed.shape}")
print(f"X_test_processed shape:  {X_test_processed.shape}")
```

**การออกแบบใน Pipeline นี้:**
- การเติมค่าว่าง (Imputation): ใช้ SimpleImputer ภายใน Pipeline ซึ่งเมื่อเรียกคำสั่ง .fit(X_train) จะจดจำค่า Median หรือ Mode จากชุด Train Set ไว้ เมื่อไปเจอค่าว่างใน Validation หรือ Test Set จะเอาค่าของ Train Set ไปเติม เพื่อไม่ให้โมเดลแอบเห็นข้อมูลเชิงสถิติของข้อมูลชุดทดสอบ   
- การเข้ารหัส (Encoding): 1. ข้อมูลหมวดหมู่จะถูกแปลงด้วย OneHotEncoder โดยมีการตั้งค่า handle_unknown='ignore' ซึ่งจำเป็นต่อการทำ OOT (Out-of-Time) Validation เพราะในข้อมูล Test (ปีล่าสุด) อาจมีหมวดหมู่ใหม่ ที่ไม่เคยเห็นใน Train Set เช่น สภาพอากาศแปลกๆ การตั้งค่านี้จะป้องกันไม่ให้เกิด Error ตอนรันโค้ด  2. ตั้งค่า sparse_output=True เพื่อประหยัดหน่วยความจำ (RAM) เนื่องจากข้อมูล Train มีขนาดกว่า 5 ล้านแถว หากแปลงแบบ Dense Array ธรรมดาอาจทำให้หน่วยความจำเต็ม (MemoryError) ได้
- การปรับสเกล (Scaling): StandardScaler จะเรียนรู้ค่า Mean และ Standard Deviation จาก Train Set เท่านั้น ทำให้ข้อมูลใหม่ที่เข้ามาถูกจัดสัดส่วนอิงจากมาตรฐานเดียวกันกับตอนที่โมเดลใช้เรียนรู้

- XGBoost (Extreme Gradient Boosting - ด้วยโหมด Histogram)

เหตุผล: เป็นโมเดลที่มีความแม่นยำสูงมากและชนะการแข่งขัน Kaggle มากมาย เมื่อเปิดใช้งานพารามิเตอร์ tree_method='hist' จะช่วยให้ XGBoost สามารถประมวลผลข้อมูล 5 ล้านแถวได้รวดเร็วใกล้เคียงกับ LightGBM เพื่อนำมาเปรียบเทียบประสิทธิภาพกัน

**กลยุทธ์การจูนพารามิเตอร์ (Hyperparameter Tuning)**
- วิธีที่เลือกใช้: Tuning ด้วย Validation Set (X_val, y_val) โดยตรง (ไม่ใช้ K-Fold Cross Validation)

**เหตุผล:** การทำ K-Fold CV (เช่น 5-Fold) บนข้อมูล 5 ล้านแถว หมายความว่าโมเดลต้องรันถึง 5 รอบต่อ 1 ชุดพารามิเตอร์ ซึ่งจะใช้เวลาและทรัพยากรคอมพิวเตอร์มากโดยไม่จำเป็น การใช้ชุด Validation Set (1,078,019 แถว) ที่แบ่งแบบ Out-of-Time ไว้แล้ว มีขนาดใหญ่และเป็นตัวแทนทางสถิติที่น่าเชื่อถือเพียงพอสำหรับการใช้ประเมิน (Evaluate) และใช้เทคนิค Early Stopping ได้ทันที ซึ่งป้องกัน Overfitting ได้ดีและประหยัดเวลากว่า

**การเลือกเมตริกวัดผล (Evaluation Metric)**

- Primary Metric: ROC-AUC
เหตุผล: เป็นเมตริกที่ทนทานต่อข้อมูลที่เสียสมดุล (Imbalanced) ช่วยวัดความสามารถของโมเดลในการ "แยกแยะ" ระหว่างคลาส 0 (ไม่รุนแรง) และ 1 (รุนแรง) ได้ดีในทุกๆ ระดับ Threshold
- Secondary Metric: F1-Score (ของคลาส 1 - รุนแรง) หรือ PR-AUC (Precision-Recall AUC)
เหตุผล: หากโจทย์ทางธุรกิจเน้นให้ความสำคัญกับการตรวจจับอุบัติเหตุรุนแรงให้เจอ (Recall สูง) โดยไม่ให้เกิด False Alarm มากเกินไป (Precision) การใช้ F1-Score จะช่วยหาจุดสมดุลที่ดีที่สุดได้

```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (roc_auc_score, f1_score, classification_report,
                             confusion_matrix, roc_curve, precision_recall_curve,
                             average_precision_score)
import xgboost as xgb

# Evaluate the final model on the test set
print("เทรน Final XGBoost Model ด้วยข้อมูล Train ทั้งหมด...")

# 1. กำหนดพารามิเตอร์จากโมเดลที่ชนะ (ดึงมาจากตารางแถวที่ 7)
best_xgb_params = {'tree_method': 'hist', 'n_estimators': 400} #
best_threshold = 0.578654 # ค่า Threshold ที่ดีที่สุดจากตาราง

# คำนวณ scale_pos_weight เพื่อจัดการ Class Imbalance
scale_weight = (y_train == 0).sum() / (y_train == 1).sum()

# เทรนโมเดล
final_model = xgb.XGBClassifier(**best_xgb_params, scale_pos_weight=scale_weight, random_state=42, n_jobs=-1)
final_model.fit(X_train_processed, y_train)

# 2. ทำนายผลบน Test Set (ทำเพียงครั้งเดียว)
print("ทำนายผลบน Test Set")
start_time = time.time()
test_preds_prob = final_model.predict_proba(X_test_processed)[:, 1]

# แปลงความน่าจะเป็นเป็นคลาส (0/1) โดยใช้ best_threshold
test_preds_class = (test_preds_prob >= best_threshold).astype(int)
eval_time = time.time() - start_time

# 3. คำนวณ Metrics บน Test Set
test_roc_auc = roc_auc_score(y_test, test_preds_prob)
test_pr_auc = average_precision_score(y_test, test_preds_prob)
test_f1 = f1_score(y_test, test_preds_class)

print("สรุปผลประเมินโมเดลบน Test Set:")
print(f"ROC-AUC   : {test_roc_auc:.4f}")
print(f"PR-AUC    : {test_pr_auc:.4f}")
print(f"F1-Score  : {test_f1:.4f} (ที่ Threshold = {best_threshold:.4f})")
print(f"เวลาทำนาย  : {eval_time:.2f} วินาที")

print("Classification Report:")
print(classification_report(y_test, test_preds_class, target_names=["ไม่รุนแรง (0)", "รุนแรง (1)"]))

# Error analysis / feature importance
fig, axes = plt.subplots(2, 2, figsize=(18, 14))

# 1. Confusion Matrix
cm = confusion_matrix(y_test, test_preds_class)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0, 0],
            xticklabels=["ไม่รุนแรง", "รุนแรง"], yticklabels=["ไม่รุนแรง", "รุนแรง"])
axes[0, 0].set_title(f'Confusion Matrix (Threshold = {best_threshold:.4f})')
axes[0, 0].set_xlabel('Predicted Label')
axes[0, 0].set_ylabel('True Label')

# 2. Feature Importance (Top 15)
try:
    feature_names = preprocessor.get_feature_names_out()
except:
    feature_names = [f"Feature_{i}" for i in range(X_train_processed.shape[1])]

importance = final_model.feature_importances_
feat_imp_df = pd.DataFrame({'Feature': feature_names, 'Importance': importance})
feat_imp_df = feat_imp_df.sort_values(by='Importance', ascending=False).head(15)

sns.barplot(x='Importance', y='Feature', data=feat_imp_df, palette='viridis', ax=axes[0, 1])
axes[0, 1].set_title('Top 15 Feature Importances (XGBoost)')
axes[0, 1].set_xlabel('Importance Score')
axes[0, 1].set_ylabel('Features')

# 3. ROC Curve
fpr, tpr, _ = roc_curve(y_test, test_preds_prob)
axes[1, 0].plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {test_roc_auc:.3f})')
axes[1, 0].plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
axes[1, 0].set_xlim([0.0, 1.0])
axes[1, 0].set_ylim([0.0, 1.05])
axes[1, 0].set_xlabel('False Positive Rate')
axes[1, 0].set_ylabel('True Positive Rate')
axes[1, 0].set_title('Receiver Operating Characteristic (ROC)')
axes[1, 0].legend(loc="lower right")

# 4. Precision-Recall Curve (เหมาะสำหรับ Imbalanced Data ที่สุด)
precision, recall, _ = precision_recall_curve(y_test, test_preds_prob)
axes[1, 1].plot(recall, precision, color='purple', lw=2, label=f'PR curve (AUC = {test_pr_auc:.3f})')
axes[1, 1].set_xlabel('Recall')
axes[1, 1].set_ylabel('Precision')
axes[1, 1].set_title('Precision-Recall Curve (PR-AUC)')
axes[1, 1].legend(loc="upper right")

plt.tight_layout()
plt.show()
```
