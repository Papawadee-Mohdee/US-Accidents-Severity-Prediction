## 1. ปัญหาและคำถามที่สนใจ
**ปัญหา/คำถามที่สนใจคืออะไร** <br>
ปัจจัยด้าน`สภาพอากาศ` `เวลา` `สถานที่` และ`ลักษณะถนน` สามารถใช้ทำนายว่าอุบัติเหตุทางถนนในสหรัฐอเมริกาจะส่ง **ผลกระทบต่อการจราจรรุนแรง** (Severity ระดับ 3–4) ได้หรือไม่ และปัจจัยใดมีความสัมพันธ์กับความรุนแรงมากที่สุด<br>

> หมายเหตุ: `Severity` ในชุดข้อมูลนี้วัด **ผลกระทบต่อการจราจร** (เช่น ความล่าช้า) ไม่ใช่ความรุนแรงของการบาดเจ็บ<br><br>

**ปัญหา/คำถามนี้มีความสำคัญอย่างไร และใครจะนำผลไปใช้** <br>
ความสำคัญ:

- อุบัติเหตุที่ส่งผลกระทบต่อการจราจรรุนแรงทำให้รถติดยาว สูญเสียเวลาและเศรษฐกิจ และเพิ่มความเสี่ยงอุบัติเหตุซ้ำ
- หากทำนายได้ว่าสถานการณ์แบบใดมีแนวโน้มรุนแรง จะจัดสรรทรัพยากรและวางมาตรการป้องกันได้ตรงจุดมากขึ้น
- ข้อมูลมีขนาดประมาณ 7.7 ล้านแถว จึงเป็นงาน Big Data ที่ต้องจัดการหน่วยความจำอย่างระมัดระวัง (ใช้ไฟล์ Parquet อ่านเฉพาะคอลัมน์ที่ใช้ และใช้ตัวอย่างข้อมูลในขั้นเลือกและจูนโมเดล) <br><br>

ผู้ใช้ประโยชน์:

- หน่วยงานดูแลความปลอดภัยทางถนนและผู้วางแผนการจราจร ใช้กำหนดจุดเสี่ยงและช่วงเวลาที่ต้องเฝ้าระวัง
- หน่วยแพทย์ฉุกเฉินและตำรวจทางหลวง ใช้จัดลำดับความสำคัญของเหตุการณ์เพื่อประเมินความต้องการทรัพยากรล่วงหน้า
- บริษัทประกันภัย ใช้ประกอบการประเมินความเสี่ยง
- แอปนำทางและผู้ขับขี่ ใช้เตือนเส้นทางที่อาจติดหนัก<br><br>

**เป้าหมาย (target) และประเภทงาน (classification / regression / unsupervised)**<br>
- **Target:** `y_severe` = 1 เมื่อ `Severity` ≥ 3 (ระดับ 3–4 = รุนแรง) และ 0 เมื่อระดับ 1–2 (ข้อมูลหลังทำความสะอาด: รุนแรง 20.4%, imbalance ratio 0.26)
- **ประเภทงาน:** Supervised Learning แบบ **Classification** (Binary Classification)
- **โมเดลที่เปรียบเทียบ:** Random Forest, XGBoost, LightGBM (scikit-learn Pipeline + ตัวอย่างข้อมูลสำหรับจูน)
- **ตัวชี้วัด:** PR-AUC เป็นตัวเลือกหลัก และรายงาน F1/Recall/Precision ของกลุ่มรุนแรง (ไม่ใช้ Accuracy อย่างเดียว เพราะข้อมูลไม่สมดุล)

```python
os.makedirs(output_dir, exist_ok=True)
chunksize = 100_000
for i, chunk in enumerate(
    pd.read_csv(input_file, chunksize=chunksize)
):
    output_file = os.path.join(
        output_dir, f"part_{i:04d}.parquet"
    )
    chunk.to_parquet(
        output_file,
        engine="pyarrow",
        compression="snappy",
        index=False
    )
    print(f"Chunk {i}: {len(chunk):,} rows saved")
    del chunk
    gc.collect()
print("Conversion completed!")
```

## 5. Data Cleaning
อธิบายทุกการตัดสินใจ **พร้อมเหตุผล** และรายงานจำนวนแถวที่เปลี่ยนแปลงในแต่ละขั้น

| ขั้นตอน | สิ่งที่ทำ | เหตุผล | แถวคงเหลือ |
|---|---|---|---|
| ... | ... | ... | ... |

```python
######################################################################
# CELL 1: ตั้งตัวบันทึกผล
######################################################################
# Handle duplicates, invalid values, outliers, inconsistent categories
cleaning_log = []

def log_step(step, action, reason, changed, rows_after):
    cleaning_log.append({"ขั้นตอน": step, "สิ่งที่ทำ": action, "เหตุผล": reason,
                         "จำนวนที่เปลี่ยน": changed, "แถวคงเหลือ": f"{rows_after:,}"})

n_rows_start = len(raw_df)
log_step("0. เริ่มต้น", "ข้อมูลหลังเลือกคอลัมน์และแก้ชนิดข้อมูล (ข้อ 4)", "-", "-", len(raw_df))
```

```python
######################################################################
# CELL 2: แถวซ้ำทุกคอลัมน์
######################################################################
# 5.1 แถวซ้ำทุกคอลัมน์ -> เก็บแถวแรก
before = len(raw_df)
raw_df = raw_df.drop_duplicates(keep="first").reset_index(drop=True)
n_removed = before - len(raw_df)
print(f"ลบแถวซ้ำทุกคอลัมน์: {n_removed:,} แถว ({n_removed / before * 100:.2f}%) | เหลือ {len(raw_df):,} แถว")
log_step("1. แถวซ้ำทุกคอลัมน์", "ลบแถวที่ซ้ำกันทุกคอลัมน์ (เก็บแถวแรก)",
         "เหตุการณ์เดียวกันถูกบันทึกซ้ำ ทำให้โมเดลเห็นข้อมูลเดิมหลายครั้ง", f"ลบ {n_removed:,} แถว", len(raw_df))
```

```python
######################################################################
# CELL 3: เวลาและพิกัดเดียวกัน
######################################################################
# 5.2 เวลาและพิกัดเดียวกันแต่ค่าอื่นต่างกัน -> ถือเป็นอุบัติเหตุเดียวกัน เก็บแถวแรก
key_cols = ["Start_Time", "Start_Lat", "Start_Lng"]
dup_any = raw_df.duplicated(subset=key_cols, keep=False)
sub = raw_df.loc[dup_any, key_cols + ["Severity"]]
g = sub.groupby(key_cols, observed=True)["Severity"].nunique()
n_groups, n_conflict = len(g), int((g > 1).sum())
print(f"กลุ่มที่เวลา+พิกัดเดียวกันแต่ค่าอื่นต่างกัน: {n_groups:,} กลุ่ม | ในนี้ Severity ขัดแย้งกัน: {n_conflict:,} กลุ่ม")
del dup_any, sub, g
gc.collect()

before = len(raw_df)
raw_df = raw_df.drop_duplicates(subset=key_cols, keep="first").reset_index(drop=True)
n_removed = before - len(raw_df)
print(f"ลบแถวที่เวลา+พิกัดซ้ำ: {n_removed:,} แถว | เหลือ {len(raw_df):,} แถว")
log_step("2. เวลา+พิกัดเดียวกัน", "เก็บแถวแรก ลบแถวที่เหลือในกลุ่ม",
         f"เวลาระดับวินาทีและพิกัดเดียวกันถือเป็นเหตุการณ์เดียวกัน (กลุ่มที่ Severity ขัดแย้ง {n_conflict:,} กลุ่ม)",
         f"ลบ {n_removed:,} แถว", len(raw_df))
```

```python
######################################################################
# CELL 4: ค่าที่เป็นไปไม่ได้
######################################################################
# 5.3 ค่าที่เป็นไปไม่ได้ -> เปลี่ยนเป็นค่าว่าง (NaN) ไม่ลบทั้งแถว
CLEAN_RULES = {
    "Temperature(F)": (-80, 135),
    "Humidity(%)": (0, 100),
    "Pressure(in)": (20, 33),      # ผ่อนจาก 25 เพราะพื้นที่สูงมีความกดต่ำกว่า (พบ 96,846 ค่าในช่วง 20-25)
    "Visibility(mi)": (0, 100),
    "Wind_Speed(mph)": (0, 250),
    "Precipitation(in)": (0, 25),
}
range_before = raw_df[list(CLEAN_RULES)].agg(["min", "max"]).T

detail, total_changed = [], 0
for col, (lo, hi) in CLEAN_RULES.items():
    bad = (raw_df[col] < lo) | (raw_df[col] > hi)
    n_bad = int(bad.sum())
    raw_df.loc[bad, col] = np.nan
    total_changed += n_bad
    detail.append({"column": col, "ช่วงที่ยอมรับ": f"{lo} ถึง {hi}", "เปลี่ยนเป็นค่าว่าง": n_bad})
display(pd.DataFrame(detail))
log_step("3. ค่าเป็นไปไม่ได้", "เปลี่ยนค่านอกช่วงที่เป็นไปได้เป็น NaN",
         "เป็นข้อผิดพลาดของการวัด/บันทึก; ไม่ลบแถวเพราะคอลัมน์อื่นยังใช้ได้", f"{total_changed:,} ช่อง", len(raw_df))
```

```python
######################################################################
# CELL 5: หมวดหมู่ที่ไม่สอดคล้อง
######################################################################
# 5.4 รวมค่าหมวดหมู่ที่หมายถึงอย่างเดียวกัน (Wind_Direction)
def remap_categories(s, func):
    """รวม/เปลี่ยนชื่อหมวดหมู่ โดยทำงานบนรายการ categories (เร็ว ไม่วนทีละแถว)"""
    cats = list(s.cat.categories)
    new_labels = [func(c) for c in cats]
    new_cats = list(dict.fromkeys(new_labels))
    pos = {c: i for i, c in enumerate(new_cats)}
    code_map = np.array([pos[x] for x in new_labels], dtype=np.int32)
    codes = s.cat.codes.to_numpy()
    new_codes = np.where(codes >= 0, code_map[np.clip(codes, 0, None)], -1)
    return pd.Series(pd.Categorical.from_codes(new_codes, categories=new_cats), index=s.index)

def wind_dir_std(x):
    m = {"north": "N", "south": "S", "east": "E", "west": "W",
         "calm": "CALM", "variable": "VAR", "var": "VAR"}
    t = str(x).strip().lower()
    return m.get(t, t.upper())

n_before = raw_df["Wind_Direction"].nunique()
raw_df["Wind_Direction"] = remap_categories(raw_df["Wind_Direction"], wind_dir_std)
n_after = raw_df["Wind_Direction"].nunique()
print(f"Wind_Direction: {n_before} ค่า -> {n_after} ค่า")
print(sorted(raw_df["Wind_Direction"].cat.categories))
log_step("4. หมวดหมู่ไม่สอดคล้อง", "รวมค่า Wind_Direction ที่ความหมายเดียวกัน (North→N, Calm→CALM, Variable→VAR ฯลฯ)",
         "ป้องกันโมเดลมองเป็นคนละกลุ่ม", f"{n_before} -> {n_after} ค่า", len(raw_df))
```

```python
######################################################################
# CELL 6: Outliers
######################################################################
# 5.5 Outliers: เทียบช่วงค่าก่อน/หลังตัดค่าเป็นไปไม่ได้ และนับ outlier ที่เหลือ (IQR)
range_after = raw_df[list(CLEAN_RULES)].agg(["min", "max"]).T
display(range_before.join(range_after, lsuffix="_ก่อน", rsuffix="_หลัง").round(2))

num_cols = ["Temperature(F)", "Humidity(%)", "Pressure(in)", "Visibility(mi)", "Wind_Speed(mph)", "Precipitation(in)"]
q = raw_df[num_cols].quantile([0.25, 0.75])
iqr = q.loc[0.75] - q.loc[0.25]
lower, upper = q.loc[0.25] - 1.5 * iqr, q.loc[0.75] + 1.5 * iqr
out = [{"column": c, "IQR": round(iqr[c], 3),
        "outliers (IQR)": int(((raw_df[c] < lower[c]) | (raw_df[c] > upper[c])).sum())} for c in num_cols]
display(pd.DataFrame(out))
log_step("5. Outliers ทางสถิติ", "เก็บไว้ (ไม่ลบ ไม่ปรับค่า)",
         "หลังตัดค่าเป็นไปไม่ได้แล้ว ที่เหลือคือสภาพอากาศสุดโต่งที่เกิดขึ้นจริง และอาจสัมพันธ์กับความรุนแรง",
         "0", len(raw_df))
```

```python
# กราฟเข้าใจง่าย: กลุ่มปัจจัยไหนเกี่ยวกับความรุนแรงมากที่สุด (เซลล์นี้คำนวณตาราง adj เอง ไม่ต้องพึ่งเซลล์อื่น ยกเว้นต้องมี clean_df)
import gc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = clean_df
df["y_severe"] = (df["Severity"] >= 3).astype("int8")           # 1 = รุนแรง (ระดับ 3-4)
year = df["Start_Time"].dt.year
hour = df["Start_Time"].dt.hour
wday = df["Start_Time"].dt.weekday
month = df["Start_Time"].dt.month
ROAD = ["Amenity", "Bump", "Crossing", "Give_Way", "Junction", "No_Exit", "Railway",
        "Roundabout", "Station", "Stop", "Traffic_Calming", "Traffic_Signal"]

def cramers_v(x, y):
    """ขนาดความสัมพันธ์ของปัจจัยหมวดหมู่ x กับ target 0/1 (0 = ไม่สัมพันธ์, 1 = สมบูรณ์) ค่าว่างนับเป็นอีกกลุ่ม"""
    codes = x.cat.codes.to_numpy() if str(x.dtype) == "category" else pd.factorize(x, use_na_sentinel=True)[0]
    idx = (codes.astype(np.int64) + 1) * 2 + y.to_numpy().astype(np.int64)
    tab = np.bincount(idx, minlength=int(idx.max()) + 2 - int(idx.max()) % 2).reshape(-1, 2).astype(float)
    tab = tab[tab.sum(axis=1) > 0]
    if tab.shape[0] < 2:
        return 0.0
    n = tab.sum()
    exp = np.outer(tab.sum(axis=1), tab.sum(axis=0)) / n
    return float(np.sqrt(((tab - exp) ** 2 / exp).sum() / n))

def cut(s, edges, labels):
    b = pd.cut(s, bins=edges, labels=labels)
    return b.cat.add_categories("Missing").fillna("Missing") if s.isna().any() else b

factors = {   # ชื่อปัจจัย: (กลุ่ม, ข้อมูล)
    "Hour of day": ("Time", hour), "Day of week": ("Time", wday), "Month": ("Time", month), "Day / night": ("Time", df["Sunrise_Sunset"]),
    "State": ("Location", df["State"]),
    "Weather condition": ("Weather", df["Weather_Condition"]),
    "Visibility": ("Weather", cut(df["Visibility(mi)"], [-0.01, 1, 3, 7, 9.99, 1e4], ["<=1", "1-3", "3-7", "7-10", "10+"])),
    "Temperature": ("Weather", cut(df["Temperature(F)"], [-200, 32, 50, 70, 90, 300], ["<32", "32-50", "50-70", "70-90", ">90"])),
    "Precipitation": ("Weather", cut(df["Precipitation(in)"], [-0.01, 0, 0.01, 0.1, 1e3], ["None", "Trace", "Light", "Heavy"])),
    "Wind speed": ("Weather", cut(df["Wind_Speed(mph)"], [-0.01, 0, 5, 10, 15, 20, 1e4], ["0", "0-5", "5-10", "10-15", "15-20", "20+"])),
}
for c in ROAD:
    factors[c] = ("Road", df[c])

# คำนวณ V "รวมทุกปี" และ "ภายในปีเดียวกัน" (ทีละปีแล้วเฉลี่ยถ่วงน้ำหนักตามจำนวนแถว) ใช้เวลาประมาณ 1-3 นาที
rows = []
for k, (g, s) in factors.items():
    v_all = cramers_v(s, df["y_severe"])
    vs, ws = [], []
    for y in sorted(year.unique()):
        m = (year == y).to_numpy()
        vs.append(cramers_v(s[m], df["y_severe"][m]))
        ws.append(m.sum())
    rows.append({"factor": k, "group": g, "V_ทั้งชุด": v_all, "V_ภายในปี": np.average(vs, weights=ws)})
adj = pd.DataFrame(rows).sort_values("V_ภายในปี", ascending=False)
display(adj.head(12).round(3))

# กราฟเข้าใจง่าย: กลุ่มปัจจัยไหนเกี่ยวกับความรุนแรงมากที่สุด
best = adj.loc[adj.groupby("group")["V_ภายในปี"].idxmax()].sort_values("V_ภายในปี")
NAMES = {"Location": "Location", "Road": "Road features", "Time": "Time", "Weather": "Weather"}
COLORS = {"Time": "#4f46e5", "Location": "#14b8a6", "Weather": "#f59e0b", "Road": "#ef4444"}

fig, ax = plt.subplots(figsize=(10, 4.8))
ax.grid(False)
yy = np.arange(len(best))
ax.barh(yy, best["V_ภายในปี"], height=0.6, color=[COLORS[g] for g in best["group"]])
xmax = best["V_ภายในปี"].max()
for i, r in enumerate(best.itertuples()):
    v_in, v_all = r[4], r[3]
    ax.text(v_in + xmax * 0.02, i + 0.08, f"{r.factor}  ({v_in:.2f})", va="center", fontsize=11, fontweight="bold")
    if v_all > 1.5 * v_in:
        ax.text(v_in + xmax * 0.02, i - 0.2, f"looked like {v_all:.2f} before removing the yearly trend", va="center", fontsize=9, color="#b45309")
ax.set_yticks(yy); ax.set_yticklabels([NAMES[g] for g in best["group"]], fontsize=12)
ax.set_xlim(0, xmax * 1.75)
ax.set_xticks([])
for sp_ in ["top", "right", "bottom"]:
    ax.spines[sp_].set_visible(False)
ax.set_title("Which group of factors relates most to severe accidents?", loc="left", fontsize=14, fontweight="bold", pad=30)
ax.text(0, 1.03, "Longer bar = stronger relationship (strongest factor in each group, compared within the same year)",
        transform=ax.transAxes, fontsize=9.5, color="#475569", va="bottom")
plt.tight_layout(); plt.show()

# คืน RAM: ตัวแปรชั่วคราวของการคำนวณ (เก็บ adj ไว้)
del factors
gc.collect()
```

1. **ตัวแปรตามไม่สมดุล** เหตุการณ์ที่ส่งผลกระทบต่อการจราจรรุนแรง (Severity ระดับ 3–4) มี 20.4% จาก 7,186,791 เหตุการณ์ (ระดับ 1/2/3/4 = 0.9/78.7/18.0/2.5%) imbalance ratio = 0.26 (ต่ำกว่า 0.3) หากทายว่า "ไม่รุนแรง" ทุกครั้งจะได้ Accuracy 79.6% แต่จับเหตุรุนแรงไม่ได้เลย (PR-AUC ของการเดา = 0.204)
   - **ผลต่อการตัดสินใจ:** ใช้ PR-AUC เลือกโมเดลและรายงาน F1/Recall/Precision แทน Accuracy, ใช้ `class_weight="balanced"` เฉพาะตอนสอน (train) และเทียบโมเดลกับ DummyClassifier

2. **สัดส่วนเหตุรุนแรงลดลงตามปีอย่างมาก** (35.7% ในปี 2018 → 2.2% ในปี 2023) ทั้งที่จำนวนบันทึกเพิ่มจาก 4.1 แสนเป็น 1.54 ล้านต่อปี (2016 → 2022) การลดลงมาจากระดับ 3 ที่หายไป (30.7% ในปี 2016 → 0.0% ในปี 2023) ขณะที่ระดับ 4 อยู่ที่ราว 1.7–3.5% ตลอด และค่าว่างของ `Precipitation` กระจุกในปีแรก ๆ (87–91% ในปี 2016–2018 แต่ไม่เกิน 5% ตั้งแต่ปี 2020) ซึ่งน่าจะสะท้อนว่าวิธีเก็บหรือให้ระดับข้อมูลเปลี่ยนไปตามเวลา (เป็นข้อสันนิษฐาน) ไม่ใช่ความปลอดภัยในโลกจริงที่เปลี่ยนไป
   - **ผลต่อการตัดสินใจ:** (ก) ไม่ใช้ปี (`year`) เป็น feature (ข) เทียบความสัมพันธ์ของปัจจัยภายในปีเดียวกัน (ค) พิจารณาไม่ใช้ตัวบอก "เคยหาย" ของ `Precipitation`/`Wind_Speed` เพราะอาจกลายเป็นตัวแทนของปี (ง) การแบ่ง train/validation/test ต้องคำนึงถึงแนวโน้มนี้ เพราะสัดส่วนเหตุรุนแรงของช่วงต้นกับท้ายต่างกันมาก และรายงานผลเทียบกับค่าเดาของแต่ละชุดหรือแยกรายปี (จ) ตัดสินใจเลือกช่วงข้อมูลที่ป้ายกำกับสม่ำเสมอ [ใส่ผลที่กลุ่มตกลง]

3. **ความสัมพันธ์ของปัจจัยกับความรุนแรง (เทียบภายในปีเดียวกัน ใช้ Cramér's V)** สถานที่ (State, 0.25) สัมพันธ์สูงสุด รองลงมาคือลักษณะถนน (Traffic_Signal 0.13, Crossing 0.12) และเวลา (เดือน 0.09, ชั่วโมง 0.08, วันในสัปดาห์ 0.07) ส่วนสภาพอากาศอ่อนที่สุด (Weather condition 0.05) โดยตัวแปรอากาศที่ดูสัมพันธ์สูงเมื่อรวมทุกปี (0.20) อ่อนลงมากเมื่อตัดแนวโน้มรายปี อย่างไรก็ตามปัจจัยเดี่ยวสัมพันธ์ระดับไม่สูง (V สูงสุดเพียง 0.25)
   - **ผลต่อการตัดสินใจ:** สร้าง feature จากทั้งสี่กลุ่ม (ชั่วโมง วันในสัปดาห์ ช่วงเร่งด่วน State พิกัด ลักษณะถนน จัดกลุ่ม `Weather_Condition`) ใช้โมเดลที่รวมหลายปัจจัยพร้อมกัน (Logistic Regression, Random Forest, HistGradientBoosting) และให้ผลโมเดลตัดสินว่าทำนายได้ดีกว่าการเดาเพียงใด ระวังตัวแปรเดือน เพราะปี 2016 เริ่มกลางเดือนมกราคมและปี 2023 มีข้อมูลแค่ 3 เดือน

**ข้อจำกัด:** ทั้งหมดเป็นความสัมพันธ์ ไม่ใช่เหตุและผล, State อาจสะท้อนวิธีบันทึกที่ต่างกันตามภูมิภาค, และข้อมูลรวมจากหลายแหล่งที่ให้ระดับ Severity ไม่เหมือนกัน

### 6.5 ตรวจสาเหตุที่สัดส่วนเหตุรุนแรงลดลงตามปี

```python
# [6.5] ตรวจสาเหตุที่สัดส่วนเหตุรุนแรงลดลงตามปี: (ก) ระดับ Severity ตามปี (ข) ค่าว่างตามปี (ค) Source (ใช้ข้อมูลดิบก่อนทำความสะอาด)
import gc
import numpy as np
import pandas as pd

yr_chk = clean_df["Start_Time"].dt.year

# (ก) สัดส่วนแต่ละระดับ Severity ตามปี: ระดับไหนที่ลดลง/เพิ่มขึ้น
t_lv = (pd.crosstab(yr_chk, clean_df["Severity"], normalize="index") * 100).round(1)
t_lv["severe (3-4)"] = t_lv[3] + t_lv[4]
print("(ก) สัดส่วนแต่ละระดับ Severity ตามปี (%):")
display(t_lv)

# (ข) ค่าว่างของตัวแปรอากาศเปลี่ยนตามปีไหม
print("\n(ข) % ค่าว่างตามปี:")
for c in ["Precipitation(in)", "Wind_Speed(mph)", "Weather_Condition"]:
    display((clean_df[c].isna().groupby(yr_chk).mean() * 100).round(1).to_frame(c).T)

# (ค) แหล่งข้อมูล (Source)
chk = dataset.to_table(columns=["Source", "Severity", "Start_Time"]).to_pandas(strings_to_categorical=True)
chk["year"] = pd.to_datetime(chk["Start_Time"], format="ISO8601", errors="coerce").dt.year
chk["severe"] = chk["Severity"] >= 3
print("\n(ค1) % เหตุรุนแรงแยกตาม Source:")
display(chk.groupby("Source", observed=True)["severe"].agg(n="size", severe_rate_pct=lambda s: s.mean() * 100).round(2))
print("(ค2) สัดส่วนของแต่ละ Source ในแต่ละปี (%):")
display((pd.crosstab(chk["year"], chk["Source"], normalize="index") * 100).round(1))
print("(ค3) % เหตุรุนแรง แยกตามปีและ Source:")
display((chk.groupby(["year", "Source"], observed=True)["severe"].mean().unstack() * 100).round(1))

del chk, yr_chk
gc.collect()
```

## 7. Feature Engineering

| Feature | สร้างจาก | เหตุผล (อ้าง EDA) | รู้ ณ เวลาทำนาย? |
|---|---|---|---|
| `hour`, `weekday`, `month` | `Start_Time` | ตัวแปรเวลามีความสัมพันธ์กับความรุนแรงเมื่อเทียบภายในปี (เดือน 0.09, ชั่วโมง 0.08, วัน 0.07) | ใช่ |
| `is_weekend` | `weekday` | แยกวันหยุดจากวันทำงาน | ใช่ |
| `is_rush_hour` | `hour` + วันทำงาน | ช่วง 7–9 น. และ 16–18 น. ในวันทำงาน | ใช่ |
| `day_night` | `Sunrise_Sunset` (ค่าว่าง = "Unknown") | กลางวัน/กลางคืนมีความสัมพันธ์เมื่อเทียบภายในปี (0.05) | ใช่ |
| `Weather_Group` | `Weather_Condition` | ลดหลายสิบ–ร้อยค่าเหลือ 7 กลุ่มที่มีข้อมูล (Clear 43.8%, Cloudy 41.3%, Rain 6.7%, Fog/Haze/Smoke 2.6%, Snow/Ice 2.3%, Thunderstorm 1.1%, Unknown 2.2%) | ใช่ (ค่าตรวจวัดจากสถานีใกล้เคียง) |
| `n_road_features` | ผลรวมของลักษณะถนน 12 ตัว | สรุปความซับซ้อนของจุดเกิดเหตุเป็นตัวเลขเดียว (aggregation) | ใช่ |
| `State`, `Start_Lat`, `Start_Lng` | คอลัมน์เดิม | สถานที่สัมพันธ์สูงสุดใน EDA (State 0.25) | ใช่ |
| ลักษณะถนน 12 ตัว | คอลัมน์เดิม | Traffic_Signal 0.13, Crossing 0.12 สูงและไม่ผันตามปี | ใช่ |
| `Temperature`, `Humidity`, `Pressure`, `Visibility`, `Wind_Direction` | คอลัมน์เดิม | องค์ประกอบของกลุ่มสภาพอากาศ | ใช่ |
| `Wind_Speed`, `Precipitation` | คอลัมน์เดิม | **ยังไม่ใช้เป็นค่าเริ่มต้น**: ว่างเกือบหมดในปีแรก ๆ (Precipitation 87–91%, Wind_Speed ~19% ในปี 2016–2018) อาจกลายเป็นตัวแทนของปี ตัดสินใจหลังเลือกช่วงข้อมูล | ใช่ |

**ไม่ใช้:** `End_Time`, `Distance(mi)`, `End_Lat/Lng`, `Description` (รู้หลังเกิดเหตุ), `ID`, `year` (ปีในอนาคตที่โมเดลไม่เคยเห็น และเป็นตัวแทนของวิธีเก็บข้อมูล), `Source` (ไม่ใช่ลักษณะของอุบัติเหตุ)

**ข้อสังเกต:** ตัวแปรเดือนอาจปนกับแนวโน้มปี เพราะปี 2016 เริ่มกลางเดือนมกราคม และปี 2023 มีข้อมูลแค่ 3 เดือน

```python
# ข้อ 7: Feature Engineering
import numpy as np
import pandas as pd

df = clean_df
SEVERE_MIN = 3
df["y_severe"] = (df["Severity"] >= SEVERE_MIN).astype("int8")      # target: 1 = รุนแรง (ระดับ 3-4)

# ---- ตัวเลือกที่ต้องตัดสินใจร่วมกับกลุ่ม ----
# Precipitation/Wind_Speed ว่างเกือบหมดในปีแรก ๆ (87-91% / 19%) ถ้าใส่เป็น feature อาจกลายเป็น "ตัวแทนของปี" แทนที่จะเป็นอากาศจริง
# ตั้ง False ไว้ก่อน เปลี่ยนเป็น True ได้ถ้าเลือกช่วงข้อมูลที่ค่าเหล่านี้ครบ (เช่น ปี 2020 เป็นต้นไป)
INCLUDE_RISKY_WEATHER = False

# ---- 1) ฟีเจอร์เวลา (จาก Start_Time) ----
t = df["Start_Time"]
df["hour"] = t.dt.hour.astype("int8")
df["weekday"] = t.dt.weekday.astype("int8")            # 0 = จันทร์
df["month"] = t.dt.month.astype("int8")
df["is_weekend"] = df["weekday"] >= 5
df["is_rush_hour"] = (~df["is_weekend"]) & df["hour"].isin([7, 8, 9, 16, 17, 18])

# ---- 2) กลางวัน/กลางคืน (ค่าว่างเป็น "Unknown") ----
df["day_night"] = df["Sunrise_Sunset"].astype("category")
if "Unknown" not in df["day_night"].cat.categories:
    df["day_night"] = df["day_night"].cat.add_categories("Unknown")
df["day_night"] = df["day_night"].fillna("Unknown")

# ---- 3) จัดกลุ่มสภาพอากาศ (Weather_Condition หลายสิบ-ร้อยค่า -> 8 กลุ่ม) ----
def remap_categories(s, func):
    """รวม/เปลี่ยนชื่อหมวดหมู่ด้วยฟังก์ชัน โดยทำงานบนรายการ categories (เร็ว ไม่วนทีละแถว)"""
    cats = list(s.cat.categories)
    new_labels = [func(c) for c in cats]
    new_cats = list(dict.fromkeys(new_labels))
    pos = {c: i for i, c in enumerate(new_cats)}
    code_map = np.array([pos[x] for x in new_labels], dtype=np.int32)
    codes = s.cat.codes.to_numpy()
    new_codes = np.where(codes >= 0, code_map[np.clip(codes, 0, None)], -1)
    return pd.Series(pd.Categorical.from_codes(new_codes, categories=new_cats), index=s.index)

def weather_group(x):
    s = str(x).lower()
    if "thunder" in s or "t-storm" in s:
        return "Thunderstorm"
    if any(k in s for k in ["snow", "sleet", "wintry", "ice", "freezing", "hail", "blizzard"]):
        return "Snow/Ice"
    if any(k in s for k in ["rain", "drizzle", "shower", "precipitation"]):
        return "Rain"
    if any(k in s for k in ["fog", "mist", "haze", "smoke", "dust", "sand", "ash"]):
        return "Fog/Haze/Smoke"
    if any(k in s for k in ["cloud", "overcast"]):
        return "Cloudy"
    if any(k in s for k in ["clear", "fair"]):
        return "Clear"
    if any(k in s for k in ["wind", "squall", "tornado", "funnel"]):
        return "Windy/Other"
    return "Other"

df["Weather_Group"] = remap_categories(df["Weather_Condition"], weather_group)
for c in ["Weather_Group", "Wind_Direction"]:
    if "Unknown" not in df[c].cat.categories:
        df[c] = df[c].cat.add_categories("Unknown")
    df[c] = df[c].fillna("Unknown")      # ค่าว่าง = กลุ่ม "Unknown"

# ---- 4) จำนวนลักษณะถนนที่พบรวมกัน (aggregation) ----
ROAD_COLS = ["Amenity", "Bump", "Crossing", "Give_Way", "Junction", "No_Exit", "Railway",
             "Roundabout", "Station", "Stop", "Traffic_Calming", "Traffic_Signal"]
df["n_road_features"] = df[ROAD_COLS].sum(axis=1).astype("int8")

# ---- รายการ feature สุดท้าย ----
CAT_COLS = ["State", "Wind_Direction", "Weather_Group", "day_night", "hour", "weekday", "month"]
NUM_COLS = ["Temperature(F)", "Humidity(%)", "Pressure(in)", "Visibility(mi)", "Start_Lat", "Start_Lng", "n_road_features"]
if INCLUDE_RISKY_WEATHER:
    NUM_COLS += ["Wind_Speed(mph)", "Precipitation(in)"]
BIN_COLS = ROAD_COLS + ["is_weekend", "is_rush_hour"]
FEATURES = CAT_COLS + NUM_COLS + BIN_COLS

# ---- ตรวจ data leakage: ห้ามมีคอลัมน์ที่รู้หลังเกิดเหตุ หรือที่ไม่ควรใช้ ----
FORBIDDEN = {"End_Time", "Distance(mi)", "End_Lat", "End_Lng", "Description", "ID", "Severity", "y_severe",
             "Source", "Start_Time", "Weather_Timestamp", "year"}
leak = FORBIDDEN.intersection(FEATURES)
assert not leak, f"พบคอลัมน์ต้องห้ามใน feature: {leak}"

# ---- ตารางสำหรับขั้นต่อไป (เก็บ Start_Time ไว้แบ่งข้อมูล และ Severity/y_severe เป็น target ไม่ใช่ feature) ----
model_df = df[FEATURES + ["y_severe", "Severity", "Start_Time"]].copy()
# ---- แสดงผลลัพธ์ ----
print(f"model_df: {model_df.shape[0]:,} แถว | features = {len(FEATURES)} "
      f"(หมวดหมู่ {len(CAT_COLS)}, ตัวเลข {len(NUM_COLS)}, 0/1 {len(BIN_COLS)}) | {model_df.memory_usage(deep=True).sum() / 1e9:.2f} GB")
print("หมวดหมู่:", CAT_COLS)
print("ตัวเลข  :", NUM_COLS)
print("0/1     :", BIN_COLS)
print("ตรวจ leakage: ผ่าน (ไม่มีคอลัมน์ต้องห้ามใน feature)")

print("\nสัดส่วนกลุ่มสภาพอากาศ (Weather_Group):")
display((df["Weather_Group"].value_counts(normalize=True) * 100).round(2).to_frame("% ของแถว"))

print("ตัวอย่าง feature ใหม่ 5 แถวแรก:")
display(model_df[["Start_Time", "hour", "weekday", "month", "is_weekend", "is_rush_hour", "day_night", "Weather_Group", "n_road_features", "y_severe"]].head())

# ตรวจความซ้ำซ้อนของตัวแปรตัวเลข (Spearman บนตัวอย่าง 300,000 แถว)
cs = df[NUM_COLS].sample(n=min(300_000, len(df)), random_state=42).corr(method="spearman")
pairs = cs.where(np.triu(np.ones(cs.shape, dtype=bool), 1)).stack()
print("คู่ตัวเลขที่สัมพันธ์กันเองสูงสุด:", f"{pairs.abs().idxmax()} |rho| = {pairs.abs().max():.2f}",
      "-> ต่ำกว่า 0.95 ไม่ต้องตัดตัวใด" if pairs.abs().max() < 0.95 else "-> ควรตัดตัวหนึ่ง")
```

## 8. Data Partitioning
กลยุทธ์การแบ่งข้อมูลที่เหมาะสมที่สุดคือ Time-Based Split (การแบ่งข้อมูลตามลำดับเวลา) หรือ Out-of-Time (OOT) Validation แทนการสุ่มแบ่งแบบปกติ (Random Split)

**การแบ่งข้อมูล (Time-Based Split)** เรียงลำดับข้อมูลตามคอลัมน์ Start_Time จากอดีตไปหาปัจจุบัน จากนั้นตัดแบ่งข้อมูลตามช่วงเวลา โดยอิงจากสัดส่วนของข้อมูลหรือจำนวนปี/เดือนที่มี

**โดยมีสัดส่วนดังนี้:**

- Training Set (70%): ใช้ข้อมูลช่วงแรกหรือข้อมูลช่วงที่เก่าที่สุด เพื่อให้โมเดลเรียนรู้รูปแบบความสัมพันธ์ของสภาพอากาศ และปัจจัยแวดล้อมที่ส่งผลต่อความรุนแรงของอุบัติเหตุ
- Validation Set (15%): ใช้ข้อมูลช่วงถัดมา สำหรับปรับจูน Hyperparameter และเลือกโมเดลที่ดีที่สุด
- Test Set (15%): ใช้ข้อมูลช่วงล่าสุด (ใหม่ที่สุด) เพื่อประเมินประสิทธิภาพขั้นสุดท้ายของโมเดลก่อนนำไปใช้งานจริง

```python
# Partition data
import pandas as pd
import numpy as np
# 1. เรียงลำดับข้อมูลตาม 'Start_Time' จากอดีตไปปัจจุบันเพื่อป้องกัน Data Leakage
model_df = model_df.sort_values('Start_Time').reset_index(drop=True)

# 2. กำหนดสัดส่วน Train 70%, Validation 15%, Test 15%
total_rows = len(model_df)
train_end = int(total_rows * 0.70)
val_end = int(total_rows * 0.85)

# 3. แบ่งข้อมูลตามลำดับเวลา
train_df = model_df.iloc[:train_end]
val_df = model_df.iloc[train_end:val_end]
test_df = model_df.iloc[val_end:]

# 4. แยก Features (X) และ Target (y)
cols_to_drop = ['y_severe', 'Severity', 'Start_Time']

X_train = train_df.drop(columns=cols_to_drop)
y_train = train_df['y_severe']

X_val = val_df.drop(columns=cols_to_drop)
y_val = val_df['y_severe']

X_test = test_df.drop(columns=cols_to_drop)
y_test = test_df['y_severe']

# 5. ตรวจสอบผลลัพธ์
print("สรุปขนาดของชุดข้อมูลหลัง Partition:")
print(f"Train Shape: {X_train.shape} | สัดส่วน: {len(X_train)/total_rows:.1%}")
print(f"Val Shape:   {X_val.shape} | สัดส่วน: {len(X_val)/total_rows:.1%}")
print(f"Test Shape:  {X_test.shape} | สัดส่วน: {len(X_test)/total_rows:.1%}")
```

## 9. Preprocessing Pipeline
การจัดการ missing values, encoding, scaling ที่ต้อง "เรียนรู้จากข้อมูล" ให้ fit จาก **train set เท่านั้น**
แล้วนำไป transform ชุดอื่น (แนะนำใช้ `sklearn.pipeline.Pipeline` / `ColumnTransformer`)

```python
# Build preprocessing pipeline
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import pandas as pd

num_cols = NUM_COLS
cat_cols = CAT_COLS
bool_cols = BIN_COLS

# 2. แก้ปัญหาชนิดข้อมูล: แปลงคอลัมน์กลุ่ม bool_cols ให้เป็นตัวเลข int (0/1)
for col in bool_cols:
    X_train[col] = X_train[col].astype(int)
    X_val[col] = X_val[col].astype(int)
    X_test[col] = X_test[col].astype(int)

# 3. สร้าง Pipeline ย่อยสำหรับแต่ละประเภทข้อมูล
# 3.1 ข้อมูลตัวเลข (Numerical)
num_pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())])

# 3.2 ข้อมูลหมวดหมู่ (Categorical)
cat_pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='most_frequent')),
    # ใช้ sparse_output=True เพื่อประหยัด RAM บนข้อมูลขนาดใหญ่
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=True)) ])

# 3.3 ข้อมูล 0/1 (Binary)
bool_pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='most_frequent'))])

# 4. รวม Pipeline ย่อยทั้งหมดเข้าด้วยกัน
preprocessor = ColumnTransformer(transformers=[('num', num_pipeline, num_cols),
        ('cat', cat_pipeline, cat_cols),('bool', bool_pipeline, bool_cols)],remainder='drop' )

# 5. Fit บน Train Set และ Transform (เรียนรู้จาก Train และนำไปประยุกต์ใช้กับทุกชุด)
X_train_processed = preprocessor.fit_transform(X_train)
X_val_processed = preprocessor.transform(X_val)
X_test_processed = preprocessor.transform(X_test)

print("สรุปขนาดของข้อมูลหลังผ่าน Preprocessing Pipeline:")
print(f"X_train_processed shape: {X_train_processed.shape}")
print(f"X_val_processed shape:   {X_val_processed.shape}")
print(f"X_test_processed shape:  {X_test_processed.shape}")
```

**การออกแบบใน Pipeline นี้:**

- การเติมค่าว่าง (Imputation): ใช้ SimpleImputer ภายใน Pipeline ซึ่งเมื่อเรียกคำสั่ง .fit(X_train) จะจดจำค่า Median หรือ Mode จากชุด Train Set ไว้ เมื่อไปเจอค่าว่างใน Validation หรือ Test Set จะเอาค่าของ Train Set ไปเติม เพื่อไม่ให้โมเดลแอบเห็นข้อมูลเชิงสถิติของข้อมูลชุดทดสอบ
- การเข้ารหัส (Encoding): 1. ข้อมูลหมวดหมู่จะถูกแปลงด้วย OneHotEncoder โดยมีการตั้งค่า handle_unknown='ignore' ซึ่งจำเป็นต่อการทำ OOT (Out-of-Time) Validation เพราะในข้อมูล Test (ปีล่าสุด) อาจมีหมวดหมู่ใหม่ ที่ไม่เคยเห็นใน Train Set เช่น สภาพอากาศแปลกๆ การตั้งค่านี้จะป้องกันไม่ให้เกิด Error ตอนรันโค้ด 2. ตั้งค่า sparse_output=True เพื่อประหยัดหน่วยความจำ (RAM) เนื่องจากข้อมูล Train มีขนาดกว่า 5 ล้านแถว หากแปลงแบบ Dense Array ธรรมดาอาจทำให้หน่วยความจำเต็ม (MemoryError) ได้
- การปรับสเกล (Scaling): StandardScaler จะเรียนรู้ค่า Mean และ Standard Deviation จาก Train Set เท่านั้น ทำให้ข้อมูลใหม่ที่เข้ามาถูกจัดสัดส่วนอิงจากมาตรฐานเดียวกันกับตอนที่โมเดลใช้เรียนรู้

## 10. Model Selection and Hyperparameter Tuning
* เลือกโมเดลอย่างน้อย 3 แบบ และอธิบายเหตุผลที่เลือก
* tuning ด้วย cross-validation หรือ validation set (ห้ามใช้ test set)
* เลือก metric ที่เหมาะกับปัญหา และอธิบายเหตุผล

### 10.0 โมเดลที่เลือก เหตุผล metric และวิธี tuning

**เหตุผลที่เลือกโมเดล:** เป้าหมายคือ Binary Classification บนข้อมูลตารางขนาด 7.2 ล้านแถว ที่ปัจจัยมีทั้งตัวเลข หมวดหมู่ และ 0/1 และความสัมพันธ์กับความรุนแรงไม่เป็นเส้นตรง (จาก EDA แต่ละปัจจัยสัมพันธ์ระดับต่ำถึงปานกลาง และ Logistic Regression ที่เพื่อนลองเป็น baseline ได้ ROC-AUC เพียง 0.657 ต่ำกว่าโมเดลต้นไม้ที่ได้ราว 0.72–0.74) จึงเลือกโมเดลแบบต้นไม้ 3 แบบที่ต่างวิธีกัน

| โมเดล | เหตุผลที่เลือก | ข้อจำกัด |
|---|---|---|
| **Random Forest** | รวมหลายต้นไม้ (bagging) จึงนิ่งและทนค่าสุดโต่ง ไม่ต้องปรับสเกล และให้ feature importance อธิบายผลได้ | เทรนช้าและใช้หน่วยความจำมาก |
| **XGBoost** | gradient boosting ที่ให้ผลดีกับข้อมูลตาราง และมีอยู่ในเนื้อหาที่เรียน | เทรนช้ากว่า LightGBM |
| **LightGBM** | boosting แบบ histogram เร็วมากและออกแบบสำหรับข้อมูลขนาดใหญ่ | ตีความยากกว่า Random Forest |

ไม่ใช้ SVM และ KNN เพราะเวลาเพิ่มขึ้นมากตามจำนวนข้อมูล (หลายแสนถึงหลายล้านแถว) และไม่ใช้ Lasso Regression เพราะใช้กับเป้าหมายต่อเนื่อง (แนวคิด L1 ใช้กับ Logistic Regression ได้ แต่โมเดลเชิงเส้นอ่อนกว่าอยู่แล้ว)

**Metric:** ใช้ **PR-AUC** เป็นตัวเลือกหลัก เพราะเหตุรุนแรงเป็นกลุ่มน้อย (9.6% ใน validation sample) การทายว่าไม่รุนแรงตลอดได้ Accuracy 90.4% ซึ่งสูงกว่า Random Forest ค่าเริ่มต้น (87.1%) จึงใช้ Accuracy ตัดสินไม่ได้ ค่าเดาของ PR-AUC เท่ากับสัดส่วนเหตุรุนแรง (0.0957) จึงรายงานเป็น "กี่เท่าของค่าเดา" และรายงาน Recall/Precision/F1 ของกลุ่มรุนแรงประกอบ

**วิธี tuning:**
- ทุกโมเดลใช้ตัวอย่างชุดเดียวกัน: train 300,000 แถว และ validation 200,000 แถว (สุ่มจาก train/validation ของการแบ่งตามเวลา)
- ชดเชยความไม่สมดุลที่ตัวโมเดลเฉพาะตอนสอน (`class_weight='balanced'` สำหรับ Random Forest และ LightGBM, `scale_pos_weight` สำหรับ XGBoost)
- ประเมินแต่ละชุดพารามิเตอร์ด้วย **validation set** (ไม่ใช้ test) เพราะข้อมูลเรียงตามเวลา การสุ่มแบ่ง fold แบบ shuffle จะปนข้อมูลอนาคตกับอดีต
- Random Forest ลองด้วย `RandomizedSearchCV` 8 ชุด ส่วน LightGBM และ XGBoost ลอง 4 ชุดต่อโมเดล (ระบุจำนวนชุดที่ลองจริงไว้ในผลรันด้านล่าง)

```python
# [10.1] คืน RAM ก่อนเริ่มข้อ 10 (ลบสำเนาข้อมูลเต็มที่ไม่ใช้แล้ว และเก็บตัวแปรเล็กที่ต้องใช้ต่อ)
import gc
import psutil
import numpy as np
import pandas as pd

def ram(msg=""):
    vm = psutil.virtual_memory()
    print(f"RAM {msg}: ใช้ {(vm.total - vm.available) / 1e9:.1f} / {vm.total / 1e9:.1f} GB")

ram("ก่อนลบ")

# เก็บเฉพาะที่ข้อ 10-11 ต้องใช้ต่อ: ปีของแถว train และ metadata ของ test (ไว้ทำ error analysis)
if "train_df" in globals():
    train_year = train_df["Start_Time"].dt.year.to_numpy()
if "test_df" in globals():
    test_meta = test_df[["State", "Weather_Group", "hour", "Start_Time"]].copy()

drop_names = ["clean_df", "df", "raw_df", "model_df", "train_df", "val_df", "test_df", "X_train", "X_val", "X_test",
              "factors", "adj", "year", "hour", "wday", "month", "rush", "dup_mask", "samp", "cand",
              "models", "model", "best_rf", "final_model"]
freed = [n for n in drop_names if n in globals()]
for n in freed:
    del globals()[n]
gc.collect()
print("ลบแล้ว:", freed)
ram("หลังลบ")

need = ["X_train_processed", "X_val_processed", "X_test_processed", "y_train", "y_val", "y_test", "preprocessor", "cat_cols", "test_meta", "train_year"]
missing = [n for n in need if n not in globals()]
print("ตัวแปรที่ข้อ 10-11 ต้องใช้:", "ครบ ✔" if not missing else f"ขาด {missing} (ให้รันข้อ 7-9 ใหม่)")
```

```python
# Model selection and tuning
# ข้อ 10: Random Forest (1/3) เทียบด้วยค่าเริ่มต้น บน Validation  (รูปแบบเดียวกับ Step 8 ในไฟล์ ML_Pipeline_Complete_Guide)
import time
import numpy as np
import pandas as pd
import scipy.sparse as sp
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, f1_score, precision_score, recall_score,
                             average_precision_score, roc_auc_score)

# train > 5 ล้านแถว Random Forest จึงเลือก "ตัวอย่าง" จาก train/validation เพื่อเลือกโมเดลและจูน (ปรับขนาดได้)
N_TRAIN_SAMPLE = 300_000
N_VAL_SAMPLE = 200_000
rng = np.random.RandomState(42)

def take(X, y, n):
    """สุ่มตัวอย่าง n แถว คืนค่า X แบบ dense (float32) และ y (รองรับทั้ง sparse และ dense)"""
    idx = np.sort(rng.choice(X.shape[0], size=min(n, X.shape[0]), replace=False))
    Xs = X[idx]
    Xs = Xs.toarray() if sp.issparse(Xs) else np.asarray(Xs)
    return Xs.astype(np.float32), y.iloc[idx].to_numpy()

X_tr_s, y_tr_s = take(X_train_processed, y_train, N_TRAIN_SAMPLE)
X_va_s, y_va_s = take(X_val_processed, y_val, N_VAL_SAMPLE)
print(f"train sample: {X_tr_s.shape} | % รุนแรง = {y_tr_s.mean() * 100:.2f}%")
print(f"val sample  : {X_va_s.shape} | % รุนแรง = {y_va_s.mean() * 100:.2f}%  (PR-AUC ของการเดา = {y_va_s.mean():.3f})")
print(f"% รุนแรงทั้งชุด: train {y_train.mean() * 100:.2f}% | val {y_val.mean() * 100:.2f}%")

# กำหนดโมเดลที่จะทดสอบ (เพิ่ม Logistic Regression / Gradient Boosting ในดิกชันนารีนี้ได้ ใช้ลูปเดียวกัน)
models = {
    'Random Forest': RandomForestClassifier(class_weight='balanced', random_state=42, n_jobs=-1),
}

# ประเมินผลบน Validation Set
results = []
for name, model in models.items():
    t0 = time.time()
    model.fit(X_tr_s, y_tr_s)
    fit_s = time.time() - t0
    y_pred = model.predict(X_va_s)
    y_prob = model.predict_proba(X_va_s)[:, 1]

    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_va_s, y_pred),
        'F1-Score (weighted)': f1_score(y_va_s, y_pred, average='weighted'),
        'F1 (รุนแรง)': f1_score(y_va_s, y_pred),
        'Recall (รุนแรง)': recall_score(y_va_s, y_pred),
        'Precision (รุนแรง)': precision_score(y_va_s, y_pred, zero_division=0),
        'PR-AUC': average_precision_score(y_va_s, y_prob),
        'ROC-AUC': roc_auc_score(y_va_s, y_prob),
        'fit_sec': round(fit_s, 1),
    })
    r = results[-1]
    print(f"{name:25s} | Accuracy: {r['Accuracy']:.4f} | F1 (weighted): {r['F1-Score (weighted)']:.4f} | PR-AUC: {r['PR-AUC']:.4f} | fit {fit_s:.0f}s")

# เรียงตาม PR-AUC (ข้อมูลไม่สมดุล จึงไม่ใช้ Accuracy/F1 weighted เป็นตัวตัดสินหลัก)
results_df = pd.DataFrame(results).sort_values('PR-AUC', ascending=False).reset_index(drop=True)
display(results_df.round(4))
print(f"ค่าเดา: PR-AUC = {y_va_s.mean():.3f} | Accuracy (ทาย 'ไม่รุนแรง' ตลอด) = {1 - y_va_s.mean():.3f}")

# คืน RAM: Random Forest ค่าเริ่มต้นที่ไม่จำกัดความลึกใช้หน่วยความจำมาก (เก็บผลไว้ใน results_df แล้ว)
import gc
del model, models
gc.collect()
```

```python
# ข้อ 10: Random Forest (2/3) Hyperparameter Tuning ด้วย RandomizedSearchCV (รูปแบบเดียวกับ Step 9 ในไฟล์ของอาจารย์)
from sklearn.model_selection import RandomizedSearchCV, PredefinedSplit

# ข้อมูลเรียงตามเวลา จึงไม่ใช้ StratifiedKFold(shuffle=True) แบบของอาจารย์ แต่ใช้ "validation set" เป็นชุดประเมิน
# (เทมเพลตอนุญาต: tuning ด้วย cross-validation หรือ validation set) -1 = ใช้สอนเท่านั้น, 0 = ชุด validation
X_cv = np.vstack([X_tr_s, X_va_s])
y_cv = np.concatenate([y_tr_s, y_va_s])
cv = PredefinedSplit(np.r_[-np.ones(len(y_tr_s)), np.zeros(len(y_va_s))])
X_tr_s, X_va_s = X_cv[:len(y_tr_s)], X_cv[len(y_tr_s):]      # ใช้ view ของ X_cv แทนสำเนา (ประหยัด RAM)

N_ITER = 8     # อาจารย์ใช้ 30 แต่ข้อมูลใหญ่ ลดลงเพื่อให้รันไหว (เพิ่มได้ถ้าเครื่องไหว)

param_grids = {
    'Random Forest': {
        'model': RandomForestClassifier(class_weight='balanced', random_state=42, n_jobs=-1),
        'params': {
            'n_estimators': [100, 200],
            'max_depth': [10, 20, None],
            'min_samples_split': [2, 10, 50],
            'min_samples_leaf': [5, 20, 50],       # ตั้งขั้นต่ำสูงกว่าของอาจารย์ (1, 2, 4) เพื่อคุมขนาดต้นไม้บนข้อมูลใหญ่
            'max_features': ['sqrt', 0.3],
        },
        'search': 'random',
    },
}

tuning_results = {}
for name in ['Random Forest']:
    cfg = param_grids[name]
    print(f'🔧 Tuning {name}...')
    search = RandomizedSearchCV(
        cfg['model'], cfg['params'], n_iter=N_ITER, cv=cv,
        scoring='average_precision',          # PR-AUC (อาจารย์ใช้ f1_weighted ซึ่งเหมาะกับข้อมูลสมดุลกว่า)
        random_state=42, n_jobs=1, refit=False
    )
    search.fit(X_cv, y_cv)
    tuning_results[name] = search
    print(f'  Best PR-AUC (validation): {search.best_score_:.4f}')
    print(f'  Best Params:  {search.best_params_}\n')

cvr = pd.DataFrame(search.cv_results_).sort_values('rank_test_score')
display(cvr[['rank_test_score', 'mean_test_score', 'mean_fit_time'] + [f'param_{k}' for k in cfg['params']]].head(5).round(4))

# ฝึกโมเดลที่จูนแล้วบน train sample แล้วเทียบกับค่าเริ่มต้น (ยังใช้ validation เท่านั้น ไม่แตะ test)
best_rf = RandomForestClassifier(class_weight='balanced', random_state=42, n_jobs=-1, **search.best_params_)
best_rf.fit(X_tr_s, y_tr_s)
p_va = best_rf.predict_proba(X_va_s)[:, 1]

from sklearn.metrics import precision_recall_curve
prec, rec, thr_all = precision_recall_curve(y_va_s, p_va)
f1_all = 2 * prec * rec / (prec + rec + 1e-12)
THRESHOLD = float(thr_all[int(np.nanargmax(f1_all[:-1]))])      # เกณฑ์ตัดสินใจที่ให้ F1 (รุนแรง) สูงสุดบน validation
pred_va = (p_va >= THRESHOLD).astype(int)

compare = pd.DataFrame([
    {'Model': 'Random Forest (default)', 'PR-AUC': results_df.loc[0, 'PR-AUC']},
    {'Model': 'Random Forest (tuned)', 'PR-AUC': average_precision_score(y_va_s, p_va),
     'F1 (รุนแรง)': f1_score(y_va_s, pred_va), 'Recall (รุนแรง)': recall_score(y_va_s, pred_va),
     'Precision (รุนแรง)': precision_score(y_va_s, pred_va, zero_division=0)},
    {'Model': 'ค่าเดา (prior)', 'PR-AUC': y_va_s.mean()},
])
display(compare.round(4))
print(f"เกณฑ์ตัดสินใจ (เลือกจาก validation) = {THRESHOLD:.3f}")

# คืน RAM
import gc
del best_rf
gc.collect()
```

```python
# [10.3] โมเดลที่ 2: LightGBM (จูนบนตัวอย่างเดียวกับ Random Forest) -- ต้องรันก่อน [10.4]
import gc
import time
import lightgbm as lgb
import xgboost as xgb
from sklearn.model_selection import ParameterSampler
from sklearn.metrics import precision_recall_curve, roc_auc_score, average_precision_score, f1_score, recall_score, precision_score

spw = (y_tr_s == 0).sum() / (y_tr_s == 1).sum()      # ชดเชยความไม่สมดุล เทียบเท่า class_weight='balanced'
N_ITER_BOOST = 4                                     # จำนวนชุดพารามิเตอร์ที่ลองต่อโมเดล (เพิ่มได้ถ้าเวลาพอ)

def build_boost(name, params):
    if name == 'LightGBM':
        return lgb.LGBMClassifier(**params, class_weight='balanced', random_state=42, n_jobs=-1, verbose=-1)
    return xgb.XGBClassifier(**params, tree_method='hist', scale_pos_weight=spw, random_state=42, n_jobs=-1)

def val_report(name, p, params, fit_s):
    """สรุปผลบน validation ด้วย PR-AUC และเกณฑ์ตัดสินใจที่ให้ F1 (รุนแรง) สูงสุด"""
    pr, rc, th = precision_recall_curve(y_va_s, p)
    f1c = 2 * pr * rc / (pr + rc + 1e-12)
    k = int(np.nanargmax(f1c[:-1]))
    pred = (p >= th[k]).astype(int)
    ap = average_precision_score(y_va_s, p)
    return {'Model': name, 'PR-AUC': ap, 'x ของค่าเดา': ap / y_va_s.mean(), 'ROC-AUC': roc_auc_score(y_va_s, p),
            'F1 (รุนแรง)': f1_score(y_va_s, pred), 'Recall (รุนแรง)': recall_score(y_va_s, pred),
            'Precision (รุนแรง)': precision_score(y_va_s, pred, zero_division=0),
            'เกณฑ์ตัดสินใจ': float(th[k]), 'Time(s)': round(fit_s, 1), 'Params': str(params)}

def tune_boost(name, space):
    log, best = [], None
    print(f"🔧 Tuning {name} ({N_ITER_BOOST} ชุด)")
    for i, params in enumerate(ParameterSampler(space, n_iter=N_ITER_BOOST, random_state=42), start=1):
        t0 = time.time()
        model = build_boost(name, params)
        model.fit(X_tr_s, y_tr_s)
        fit_s = time.time() - t0
        row = val_report(name, model.predict_proba(X_va_s)[:, 1], params, fit_s)
        log.append(row)
        print(f"  [{i}/{N_ITER_BOOST}] PR-AUC = {row['PR-AUC']:.4f} | ROC-AUC = {row['ROC-AUC']:.4f} | {fit_s:.0f}s | {params}")
        if best is None or row['PR-AUC'] > best[1]['PR-AUC']:
            best = (params, row)
        del model
        gc.collect()
    return log, best

lgbm_space = {'n_estimators': [200, 400], 'learning_rate': [0.05, 0.1], 'num_leaves': [31, 63], 'max_depth': [5, 7]}
lgbm_log, lgbm_best = tune_boost('LightGBM', lgbm_space)
display(pd.DataFrame(lgbm_log).drop(columns=['Params']).round(4))
print(f"\n🏆 LightGBM ที่ดีที่สุด: PR-AUC = {lgbm_best[1]['PR-AUC']:.4f} ({lgbm_best[1]['x ของค่าเดา']:.2f} เท่าของค่าเดา) | Params = {lgbm_best[0]}")
```

```python
# [10.4] โมเดลที่ 3: XGBoost (จูนบนตัวอย่างเดียวกัน) -- ต้องรัน [10.3] มาก่อน
xgb_space = {'n_estimators': [200, 400], 'learning_rate': [0.05, 0.1], 'max_depth': [5, 7]}
xgb_log, xgb_best = tune_boost('XGBoost', xgb_space)
display(pd.DataFrame(xgb_log).drop(columns=['Params']).round(4))
print(f"\n🏆 XGBoost ที่ดีที่สุด: PR-AUC = {xgb_best[1]['PR-AUC']:.4f} ({xgb_best[1]['x ของค่าเดา']:.2f} เท่าของค่าเดา) | Params = {xgb_best[0]}")
```

```python
# [11-3] Error Analysis บน Test: โมเดลพลาดกับกลุ่มไหน
res = test_meta.reset_index(drop=True).copy()
res["y"], res["pred"] = y_true_test, y_pred_test
res["time_of_day"] = pd.cut(res["hour"].astype(int), bins=[-1, 5, 9, 15, 19, 23],
                            labels=["night 0-5", "morning 6-9", "midday 10-15", "evening 16-19", "late 20-23"])
res["quarter"] = res["Start_Time"].dt.to_period("Q").astype(str)

def group_metrics(key, min_n=5000):
    rows = []
    for k, g in res.groupby(key, observed=True):
        if len(g) < min_n:
            continue
        yt, pr = g["y"].to_numpy(), g["pred"].to_numpy()
        tp_ = int(((yt == 1) & (pr == 1)).sum()); fn_ = int(((yt == 1) & (pr == 0)).sum()); fp_ = int(((yt == 0) & (pr == 1)).sum())
        rec = tp_ / max(tp_ + fn_, 1); pre = tp_ / max(tp_ + fp_, 1)
        rows.append({"group": k, "n": len(g), "% รุนแรงจริง": round(yt.mean() * 100, 2), "Recall": round(rec, 3),
                     "Precision": round(pre, 3), "F1": round(2 * rec * pre / max(rec + pre, 1e-12), 3)})
    return pd.DataFrame(rows)

err_tables = {}
print(f"ภาพรวม test: Recall = {test_metrics['Recall']:.3f} | Precision = {test_metrics['Precision']:.3f}")
for key, title in [("time_of_day", "ช่วงเวลาของวัน"), ("Weather_Group", "กลุ่มสภาพอากาศ"), ("quarter", "ไตรมาส (ดูแนวโน้มตามเวลา)"), ("State", "รัฐ (10 รัฐที่มีข้อมูลมากสุด)")]:
    tbl = group_metrics(key)
    if key == "State":
        tbl = tbl.sort_values("n", ascending=False).head(10)
    err_tables[key] = tbl
    print(f"\n--- {title} ---")
    display(tbl.reset_index(drop=True))
```

```python
# [11-4] Feature Importance ของโมเดลที่เลือก (รวมคอลัมน์ One-Hot กลับเป็นชื่อ feature เดิม)
names = preprocessor.get_feature_names_out()

def original_name(n):
    if n.startswith('cat__'):
        for c in cat_cols:
            if n.startswith(f'cat__{c}_'):
                return c
    return n.split('__', 1)[1]

imp = np.asarray(final_model.feature_importances_, dtype=float)
fi_df = (pd.DataFrame({'Feature': [original_name(n) for n in names], 'Importance': imp / imp.sum()})
         .groupby('Feature', as_index=False)['Importance'].sum()
         .sort_values('Importance', ascending=False).reset_index(drop=True))

plot_fi = fi_df.head(15).sort_values('Importance')
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(plot_fi['Feature'], plot_fi['Importance'], color='#3498db', edgecolor='white')
ax.set_xlabel('Importance (sums to 1 over all features)')
ax.set_title(f'{FINAL_NAME} — Feature Importance (Top 15)')
ax.grid(axis='x', alpha=0.3)
plt.tight_layout(); plt.show()

print('💡 Top 5 features ที่สำคัญที่สุด:')
for idx, row in fi_df.head(5).iterrows():
    print(f"   {idx + 1}. {row['Feature']} (importance = {row['Importance']:.4f})")
print("หมายเหตุ: importance นี้บอกว่าโมเดล 'ใช้' feature ไหนมาก ไม่ใช่เหตุและผล")

# คืน RAM หลังจบข้อ 11 (เก็บ test_metrics, err_tables, fi_df ไว้ใช้ในข้อ 12-13)
del final_model, y_prob_test
gc.collect()
```

## 12. ทำไมโมเดลที่สร้างจึงเหมาะสม (Model Justification)
ตอบโดย **อ้างอิงหลักฐานจากผลการวิเคราะห์ (ตัวเลข ตาราง กราฟ)** ไม่ใช่ความเห็น เช่น "accuracy สูง จึงเหมาะสม" ถือว่าไม่เพียงพอ

ตารางหลักฐานของแต่ละประเด็นสร้างจากผลรันจริงในเซลล์ `[12-3]` ด้านล่าง ประเด็นที่ตอบมี 8 ข้อ:

1. **ความเหมาะกับปัญหาและข้อมูล:** โมเดลเหมาะกับชนิดของ target ขนาดข้อมูล และลักษณะ feature อย่างไร
2. **Metric:** metric ที่ใช้สะท้อนสิ่งที่ผู้ใช้ต้องการจริงหรือไม่ ความผิดพลาดแบบไหน (FN vs FP) เสียหายมากกว่ากัน
3. **ดีกว่าการทำนายแบบง่าย:** เทียบกับการเดาแบบง่ายที่สุด (Dummy) มากน้อยเพียงใด
4. **Generalization:** ผลบน train, cross-validation/validation และ test ใกล้เคียงกันหรือไม่ มี overfitting หรือไม่
5. **ความเสถียร:** ผลแต่ละ fold ของ cross-validation แปรปรวนมากหรือไม่
6. **ความสมเหตุสมผลของโมเดล:** feature ที่สำคัญสอดคล้องกับ EDA และความรู้ในโดเมนหรือไม่ (ผลที่ดีเกินจริงอาจเป็นสัญญาณของ data leakage)
7. **ข้อผิดพลาดที่ยอมรับได้:** โมเดลผิดพลาดกับข้อมูลกลุ่มไหน ยอมรับได้สำหรับการนำไปใช้หรือไม่
8. **ความซับซ้อนคุ้มค่า:** ถ้าโมเดลซับซ้อนกว่าได้ผลดีขึ้นเพียงเล็กน้อย ควรเลือกโมเดลที่ง่ายกว่าหรือไม่ เพราะอะไร

```python
# [12-1] ตัวเลข train และ cross-validation (TimeSeriesSplit) ของโมเดลที่เลือก (ยังไม่ใช้ test) -- ใช้เวลาสักพัก
import gc
import time
from sklearn.model_selection import TimeSeriesSplit

N_CV = 150_000                                   # ตัวอย่างย่อยสำหรับ CV (ประหยัดเวลา/RAM) สุ่มแล้วเรียงตามเวลา
rng_cv = np.random.RandomState(1)
sub = np.sort(rng_cv.choice(len(y_tr_s), size=min(N_CV, len(y_tr_s)), replace=False))
X_cvsub, y_cvsub = X_tr_s[sub], y_tr_s[sub]

# (1) PR-AUC บน train sample และ validation sample (ดูว่าโมเดลเรียนรู้ข้อมูลสอนมากเกินพอดีไหม)
t0 = time.time()
m = make_final(FINAL_NAME, FINAL_PARAMS)
m.fit(X_tr_s, y_tr_s)
train_ap = average_precision_score(y_tr_s, m.predict_proba(X_tr_s)[:, 1])
val_ap = average_precision_score(y_va_s, m.predict_proba(X_va_s)[:, 1])
del m
gc.collect()
print(f"Train PR-AUC = {train_ap:.4f} | Validation PR-AUC = {val_ap:.4f} | ช่องว่าง = {train_ap - val_ap:+.4f} ({time.time() - t0:.0f}s)")

# (2) Cross-validation แบบเรียงตามเวลา 3 รอบ (สอนจากช่วงต้น สอบกับช่วงถัดไป)
cv_aps, cv_lifts = [], []
for k, (a, b) in enumerate(TimeSeriesSplit(n_splits=3).split(X_cvsub), start=1):
    m = make_final(FINAL_NAME, FINAL_PARAMS)
    m.fit(X_cvsub[a], y_cvsub[a])
    ap = average_precision_score(y_cvsub[b], m.predict_proba(X_cvsub[b])[:, 1])
    base = y_cvsub[b].mean()
    cv_aps.append(ap); cv_lifts.append(ap / base)
    print(f"fold {k}: PR-AUC = {ap:.4f} | ค่าเดาของรอบนี้ = {base:.4f} | ดีกว่าเดา {ap / base:.2f} เท่า")
    del m
    gc.collect()
cv_text = f"{np.mean(cv_aps):.4f} ± {np.std(cv_aps):.4f}"
print(f"CV PR-AUC (mean ± SD) = {cv_text} | ดีกว่าเดาเฉลี่ย {np.mean(cv_lifts):.2f} เท่า")
del X_cvsub
gc.collect()
```

```python
# [12-2] ตารางเปรียบเทียบ train / CV / validation / test และเทียบกับการทำนายแบบง่าย (Dummy)
from sklearn.dummy import DummyClassifier

perf_rows = []
for _, r in compare_df.iterrows():
    is_final = r['Model'] == FINAL_NAME
    perf_rows.append({
        'Model': r['Model'] + (' (เลือก)' if is_final else ''),
        'Train PR-AUC': round(train_ap, 4) if is_final else '-',
        'CV PR-AUC (mean ± SD)': cv_text if is_final else '-',
        'Validation PR-AUC': round(r['PR-AUC'], 4),
        'Test PR-AUC': round(test_metrics['PR-AUC'], 4) if is_final else 'ไม่ประเมิน (เก็บ test ไว้ให้โมเดลที่เลือกเท่านั้น)',
    })
perf_df = pd.DataFrame(perf_rows)
display(perf_df)
print("หมายเหตุ: โมเดลที่ไม่ได้เลือกวัดบน validation เท่านั้น (เทียบกันด้วยตัวอย่างและ metric เดียวกันในข้อ 10) ตัวเลข train/CV คำนวณเฉพาะโมเดลที่เลือก")

dummy_rows = []
for strat in ['prior', 'stratified']:
    d = DummyClassifier(strategy=strat, random_state=42).fit(np.zeros((len(y_tr_s), 1)), y_tr_s)
    Xd = np.zeros((len(y_true_test), 1))
    pr_ = d.predict(Xd)
    dummy_rows.append({'Model': f'Dummy ({strat})', 'PR-AUC': average_precision_score(y_true_test, d.predict_proba(Xd)[:, 1]),
                       'F1 (รุนแรง)': f1_score(y_true_test, pr_, zero_division=0), 'Recall': recall_score(y_true_test, pr_),
                       'Precision': precision_score(y_true_test, pr_, zero_division=0)})
dummy_rows.append({'Model': f'{FINAL_NAME} (เลือก)', 'PR-AUC': test_metrics['PR-AUC'], 'F1 (รุนแรง)': test_metrics['F1'],
                   'Recall': test_metrics['Recall'], 'Precision': test_metrics['Precision']})
dummy_df = pd.DataFrame(dummy_rows)
print("\nเทียบกับการทำนายแบบง่ายบน Test:")
display(dummy_df.round(4))
print(f"PR-AUC ของโมเดล = {test_metrics['PR-AUC']:.4f} เทียบกับการเดา {test_metrics['prevalence']:.4f} (ดีกว่า {test_metrics['PR-AUC'] / test_metrics['prevalence']:.2f} เท่า)")
```

```python
# [12-3] ตารางเหตุผล 8 ประเด็นของข้อ 12 (ตัวเลขดึงจากผลรันจริงอัตโนมัติ)
pd.set_option("display.max_colwidth", None)

lift_t = test_metrics['PR-AUC'] / test_metrics['prevalence']
top5_feat = ", ".join(fi_df["Feature"].head(5))
worst = None
for key in ["time_of_day", "Weather_Group", "State"]:
    t_ = err_tables.get(key)
    if t_ is not None and len(t_):
        r_ = t_.sort_values("Recall").iloc[0]
        if worst is None or r_["Recall"] < worst[2]:
            worst = (key, r_["group"], r_["Recall"])
second = compare_df.loc[1]
d_prior = dummy_df.loc[dummy_df['Model'] == 'Dummy (prior)'].iloc[0]
d_strat = dummy_df.loc[dummy_df['Model'] == 'Dummy (stratified)'].iloc[0]
rf_time = compare_df.loc[compare_df['Model'] == 'Random Forest', 'Time(s)'].iloc[0]
lg_time = compare_df.loc[compare_df['Model'] == 'LightGBM', 'Time(s)'].iloc[0]

just = pd.DataFrame([
    ["ความเหมาะกับปัญหาและข้อมูล",
     f"งาน Binary Classification บนข้อมูลตาราง {len(y_train) + len(y_val) + len(y_test):,} แถว ที่มีทั้งตัวเลข/หมวดหมู่/0-1 และความสัมพันธ์ไม่เป็นเส้นตรง (Logistic Regression ที่ลองได้ ROC-AUC ต่ำกว่าโมเดลต้นไม้ชัดเจน); โมเดลที่เลือก = {FINAL_NAME}; ใช้ class weight และ Pipeline จัดการข้อมูลไม่สมดุลและค่าว่าง"],
    ["Metric",
     f"ใช้ PR-AUC เพราะเหตุรุนแรงมีเพียง {test_metrics['prevalence'] * 100:.1f}% ใน test (Accuracy หลอกตา); ที่เกณฑ์ {FINAL_THRESHOLD:.3f} ได้ Recall = {test_metrics['Recall']:.3f}, Precision = {test_metrics['Precision']:.3f}; FN (พลาดเหตุรุนแรง) = {fn:,} ราย, FP (เตือนผิด) = {fp:,} ราย"],
    ["ดีกว่าการทำนายแบบง่าย",
     f"PR-AUC test = {test_metrics['PR-AUC']:.4f} เทียบกับการเดา {test_metrics['prevalence']:.4f} (ดีกว่า {lift_t:.2f} เท่า); F1 = {test_metrics['F1']:.3f} เทียบกับ Dummy stratified {d_strat['F1 (รุนแรง)']:.3f} และ Dummy prior {d_prior['F1 (รุนแรง)']:.3f}"],
    ["Generalization",
     f"PR-AUC: train = {train_ap:.4f}, CV = {cv_text}, validation = {val_ap:.4f}, test = {test_metrics['PR-AUC']:.4f}; ช่องว่าง train-validation = {train_ap - val_ap:+.4f}, validation-test = {val_ap - test_metrics['PR-AUC']:+.4f} (ช่องว่างใหญ่ = overfit หรือข้อมูลเปลี่ยนตามเวลา; % รุนแรง train {y_train.mean() * 100:.1f}% / val {y_val.mean() * 100:.1f}% / test {y_test.mean() * 100:.1f}%)"],
    ["ความเสถียร",
     f"PR-AUC ของแต่ละ fold (TimeSeriesSplit) = {np.round(cv_aps, 4).tolist()}; SD = {np.std(cv_aps):.4f}"],
    ["ความสมเหตุสมผลของโมเดล",
     f"feature สำคัญ 5 อันดับแรก: {top5_feat}; ให้เทียบกับ EDA ข้อ 6 และความรู้ในโดเมน; " + ("⚠️ PR-AUC สูงมากผิดปกติ ต้องไล่ตรวจ data leakage" if test_metrics['PR-AUC'] > 0.9 else "คะแนนไม่สูงเกินจริงจนน่าสงสัยว่ามี leakage")],
    ["ข้อผิดพลาดที่ยอมรับได้",
     (f"กลุ่มที่ Recall ต่ำสุด: {worst[0]} = {worst[1]} (Recall {worst[2]:.3f}) เทียบกับ Recall รวม {test_metrics['Recall']:.3f}; " if worst else "")
     + "ถ้าใช้เป็นสัญญาณช่วยจัดลำดับความสำคัญ ยอมรับการเตือนผิดได้ แต่ไม่ควรใช้ตัดสินใจเดี่ยว ๆ"],
    ["ความซับซ้อนคุ้มค่า",
     f"อันดับ 2 บน validation คือ {second['Model']} (PR-AUC {second['PR-AUC']:.4f}) ต่างจาก {FINAL_NAME} {compare_df.loc[0, 'PR-AUC'] - second['PR-AUC']:.4f}; เวลาฝึกชุดที่ดีที่สุด Random Forest {rf_time:.0f} วินาที เทียบกับ LightGBM {lg_time:.0f} วินาที"],
], columns=["ประเด็น", "หลักฐาน (ตัวเลขจากผลรัน)"])
display(just)
```
